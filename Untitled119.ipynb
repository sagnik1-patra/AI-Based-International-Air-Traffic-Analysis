```python
import json
import pickle
import warnings
from pathlib import Path

import h5py
import joblib
import yaml
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
from sklearn.impute import SimpleImputer
from sklearn.linear_model import LinearRegression, Ridge
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import (
    RandomForestRegressor,
    GradientBoostingRegressor
)
from sklearn.multioutput import MultiOutputRegressor
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

warnings.filterwarnings("ignore")

# ============================================================
# STEP 1: PROJECT CONFIGURATION
# ============================================================

ROOT = Path(
    r"C:\Users\sagni\Downloads"
    r"\AI-Based International Air Traffic Analysis"
)

DATA_PATH = ROOT / "IntlTrafficData.csv"

ROOT.mkdir(parents=True, exist_ok=True)

RANDOM_STATE = 42
rng = np.random.default_rng(RANDOM_STATE)

sns.set_theme(style="whitegrid")

plt.rcParams.update({
    "figure.figsize": (12, 6),
    "font.size": 11,
    "savefig.dpi": 300
})

print("=" * 70)
print("PSO INTERNATIONAL AIR TRAFFIC ANALYSIS")
print("=" * 70)

print("\nProject directory:", ROOT)

# ============================================================
# STEP 2: FUNCTION TO SAVE AND DISPLAY GRAPHS
# ============================================================

def save_plot(filename):

    path = ROOT / filename

    plt.tight_layout()

    plt.savefig(
        path,
        dpi=300,
        bbox_inches="tight"
    )

    print("[SAVED]", filename)

    plt.show()
    plt.close()

# ============================================================
# STEP 3: LOAD DATASET
# ============================================================

if not DATA_PATH.exists():
    raise FileNotFoundError(
        f"Dataset not found: {DATA_PATH}"
    )

df = pd.read_csv(DATA_PATH)

print("\nDataset loaded successfully!")
print("Dataset shape:", df.shape)

print("\nFirst five records:")
print(df.head())

# ============================================================
# STEP 4: CLEAN DATASET
# ============================================================

df.columns = (
    df.columns
    .str.strip()
    .str.upper()
    .str.replace(r"\s+", "_", regex=True)
)

required_columns = [
    "YEAR",
    "OPERATOR",
    "PAX_TOTAL",
    "FREIGHT_TOTAL"
]

missing_columns = [
    col for col in required_columns
    if col not in df.columns
]

if missing_columns:
    raise ValueError(
        f"Missing required columns: {missing_columns}"
    )

df["YEAR"] = df["YEAR"].astype(str).str.strip()

df["OPERATOR"] = (
    df["OPERATOR"]
    .astype(str)
    .str.strip()
    .str.upper()
)

# Extract starting year from financial year.
# Example: 2010-11 -> 2010

df["YEAR_START"] = pd.to_numeric(
    df["YEAR"].str.extract(r"(\d{4})")[0],
    errors="coerce"
)

numeric_columns = [
    "PAX_TO",
    "PAX_FROM",
    "PAX_TOTAL",
    "FREIGHT_TO",
    "FREIGHT_FROM",
    "FREIGHT_TOTAL"
]

for col in numeric_columns:

    if col in df.columns:

        df[col] = pd.to_numeric(
            df[col].astype(str).str.replace(
                ",", "", regex=False
            ),
            errors="coerce"
        )

df = df.dropna(
    subset=[
        "YEAR_START",
        "OPERATOR",
        "PAX_TOTAL",
        "FREIGHT_TOTAL"
    ]
).copy()

df["YEAR_START"] = df["YEAR_START"].astype(int)

df = df.sort_values(
    ["YEAR_START", "OPERATOR"]
).reset_index(drop=True)

df.to_csv(
    ROOT / "pso_processed_data.csv",
    index=False
)

print("\nCleaned dataset:")
print(df.head())

print("\nMissing values:")
print(df.isnull().sum())

# ============================================================
# STEP 5: CORRELATION HEATMAP
# ============================================================

print("\nGenerating PSO correlation heatmap...")

heatmap_columns = [
    col for col in numeric_columns
    if col in df.columns
]

heatmap_data = df[
    heatmap_columns + ["YEAR_START"]
].select_dtypes(include=[np.number])

plt.figure(figsize=(12, 9))

sns.heatmap(
    heatmap_data.corr(),
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    linewidths=0.5,
    square=True
)

plt.title(
    "PSO - International Air Traffic Correlation Heatmap",
    fontsize=15,
    fontweight="bold"
)

save_plot("pso_heatmap.png")

# ============================================================
# STEP 6: SELECT FEATURES AND TARGETS
# ============================================================

FEATURES = [
    "YEAR_START",
    "OPERATOR"
]

TARGETS = [
    "PAX_TOTAL",
    "FREIGHT_TOTAL"
]

# Avoid target leakage:
# PAX_TO and PAX_FROM are not used to
# predict their sum PAX_TOTAL.

X = df[FEATURES].copy()
y = df[TARGETS].copy()

print("\nFeatures:", FEATURES)
print("Targets:", TARGETS)

# ============================================================
# STEP 7: CHRONOLOGICAL TRAIN/TEST SPLIT
# ============================================================

years = sorted(
    df["YEAR_START"].unique()
)

if len(years) < 4:
    raise ValueError(
        "At least four distinct years are required "
        "for this PSO validation setup."
    )

split_index = max(
    3,
    int(len(years) * 0.70)
)

split_index = min(
    split_index,
    len(years) - 1
)

train_years = years[:split_index]
test_years = years[split_index:]

train_mask = df["YEAR_START"].isin(train_years)
test_mask = df["YEAR_START"].isin(test_years)

X_train = X.loc[train_mask].copy()
X_test = X.loc[test_mask].copy()

y_train = y.loc[train_mask].copy()
y_test = y.loc[test_mask].copy()

print("\nTraining years:", train_years)
print("Testing years:", test_years)

print("Training samples:", len(X_train))
print("Testing samples:", len(X_test))

# ============================================================
# STEP 8: CREATE PREPROCESSOR
# ============================================================

def make_preprocessor():

    numeric_pipeline = Pipeline([
        (
            "imputer",
            SimpleImputer(strategy="median")
        )
    ])

    categorical_pipeline = Pipeline([
        (
            "imputer",
            SimpleImputer(strategy="most_frequent")
        ),
        (
            "encoder",
            OneHotEncoder(
                handle_unknown="ignore"
            )
        )
    ])

    return ColumnTransformer([
        (
            "numeric",
            numeric_pipeline,
            ["YEAR_START"]
        ),
        (
            "categorical",
            categorical_pipeline,
            ["OPERATOR"]
        )
    ])

def make_pipeline(estimator):

    return Pipeline([
        (
            "preprocessor",
            make_preprocessor()
        ),
        (
            "model",
            estimator
        )
    ])

# ============================================================
# STEP 9: EVALUATION FUNCTIONS
# ============================================================

def evaluate_predictions(y_true, y_pred):

    actual = np.asarray(
        y_true,
        dtype=float
    )

    predicted = np.maximum(
        np.asarray(y_pred, dtype=float),
        0
    )

    passenger_mae = mean_absolute_error(
        actual[:, 0],
        predicted[:, 0]
    )

    freight_mae = mean_absolute_error(
        actual[:, 1],
        predicted[:, 1]
    )

    passenger_rmse = np.sqrt(
        mean_squared_error(
            actual[:, 0],
            predicted[:, 0]
        )
    )

    freight_rmse = np.sqrt(
        mean_squared_error(
            actual[:, 1],
            predicted[:, 1]
        )
    )

    passenger_r2 = r2_score(
        actual[:, 0],
        predicted[:, 0]
    )

    freight_r2 = r2_score(
        actual[:, 1],
        predicted[:, 1]
    )

    return {
        "Passenger_MAE": float(passenger_mae),
        "Freight_MAE": float(freight_mae),
        "Passenger_RMSE": float(passenger_rmse),
        "Freight_RMSE": float(freight_rmse),
        "Passenger_R2": float(passenger_r2),
        "Freight_R2": float(freight_r2)
    }

def normalized_mae(y_true, y_pred, scales):

    actual = np.asarray(
        y_true,
        dtype=float
    )

    predicted = np.maximum(
        np.asarray(y_pred, dtype=float),
        0
    )

    errors = np.mean(
        np.abs(actual - predicted),
        axis=0
    )

    return float(
        np.mean(errors / scales)
    )

# ============================================================
# STEP 10: PSO HYPERPARAMETER SEARCH SPACE
# ============================================================

# Particle dimensions:
#
# 0 = n_estimators
# 1 = max_depth
# 2 = min_samples_split
# 3 = min_samples_leaf
# 4 = max_features

LOWER_BOUNDS = np.array(
    [30, 2, 2, 1, 0.4],
    dtype=float
)

UPPER_BOUNDS = np.array(
    [150, 8, 8, 4, 1.0],
    dtype=float
)

DIMENSIONS = len(LOWER_BOUNDS)

def decode_particle(position):

    position = np.clip(
        np.asarray(position, dtype=float),
        LOWER_BOUNDS,
        UPPER_BOUNDS
    )

    return {
        "n_estimators": int(round(position[0])),
        "max_depth": int(round(position[1])),
        "min_samples_split": int(round(position[2])),
        "min_samples_leaf": int(round(position[3])),
        "max_features": float(position[4])
    }

def make_pso_estimator(params):

    return RandomForestRegressor(
        **params,
        random_state=RANDOM_STATE,
        n_jobs=-1
    )

# ============================================================
# STEP 11: EXPANDING-WINDOW VALIDATION
# ============================================================

# PSO must not optimize on the test set.
# We create validation folds using training
# years only.

def build_validation_folds():

    unique_years = sorted(
        X_train["YEAR_START"].unique()
    )

    folds = []

    for i in range(2, len(unique_years)):

        fit_years = unique_years[:i]
        validation_year = unique_years[i]

        fit_mask = X_train[
            "YEAR_START"
        ].isin(fit_years)

        validation_mask = (
            X_train["YEAR_START"]
            == validation_year
        )

        folds.append((
            X_train.loc[fit_mask],
            y_train.loc[fit_mask],
            X_train.loc[validation_mask],
            y_train.loc[validation_mask]
        ))

    if not folds:
        raise ValueError(
            "Insufficient training years "
            "for time-aware validation."
        )

    return folds

validation_folds = build_validation_folds()

# Use training-set means to normalize
# passenger and freight errors.

target_scales = np.maximum(
    np.abs(
        y_train.mean(axis=0).to_numpy(
            dtype=float
        )
    ),
    1.0
)

fitness_cache = {}

# ============================================================
# STEP 12: PSO FITNESS FUNCTION
# ============================================================

def pso_fitness(position):

    params = decode_particle(position)

    key = tuple(params.items())

    if key in fitness_cache:
        return fitness_cache[key]

    fold_errors = []

    for (
        X_fit,
        y_fit,
        X_val,
        y_val
    ) in validation_folds:

        model = make_pipeline(
            make_pso_estimator(params)
        )

        model.fit(X_fit, y_fit)

        predictions = model.predict(X_val)

        error = normalized_mae(
            y_val,
            predictions,
            target_scales
        )

        fold_errors.append(error)

    fitness = float(
        np.mean(fold_errors)
    )

    fitness_cache[key] = fitness

    return fitness

# ============================================================
# STEP 13: PSO CONFIGURATION
# ============================================================

NUM_PARTICLES = 10
NUM_ITERATIONS = 12

INERTIA_WEIGHT = 0.7

COGNITIVE_COEFFICIENT = 1.5
SOCIAL_COEFFICIENT = 1.5

# Initialize particle positions

positions = rng.uniform(
    LOWER_BOUNDS,
    UPPER_BOUNDS,
    size=(NUM_PARTICLES, DIMENSIONS)
)

# Initialize particle velocities

velocity_limits = (
    UPPER_BOUNDS - LOWER_BOUNDS
) * 0.20

velocities = rng.uniform(
    -velocity_limits,
    velocity_limits,
    size=(NUM_PARTICLES, DIMENSIONS)
)

# Personal best positions

personal_best_positions = positions.copy()

personal_best_scores = np.array([
    pso_fitness(position)
    for position in positions
])

# Global best

best_index = int(
    np.argmin(personal_best_scores)
)

global_best_position = (
    personal_best_positions[best_index].copy()
)

global_best_score = float(
    personal_best_scores[best_index]
)

optimization_history = []

# ============================================================
# STEP 14: RUN PARTICLE SWARM OPTIMIZATION
# ============================================================

print("\nStarting PSO optimization...")

for iteration in range(NUM_ITERATIONS):

    current_scores = np.array([
        pso_fitness(position)
        for position in positions
    ])

    # Update personal best

    for i in range(NUM_PARTICLES):

        if (
            current_scores[i]
            < personal_best_scores[i]
        ):

            personal_best_scores[i] = (
                current_scores[i]
            )

            personal_best_positions[i] = (
                positions[i].copy()
            )

    # Update global best

    best_index = int(
        np.argmin(personal_best_scores)
    )

    if (
        personal_best_scores[best_index]
        < global_best_score
    ):

        global_best_score = float(
            personal_best_scores[best_index]
        )

        global_best_position = (
            personal_best_positions[
                best_index
            ].copy()
        )

    optimization_history.append({
        "Iteration": iteration + 1,
        "Global_Best_Fitness": global_best_score,
        "Mean_Particle_Fitness": float(
            np.mean(current_scores)
        ),
        "Worst_Particle_Fitness": float(
            np.max(current_scores)
        )
    })

    print(
        f"Iteration {iteration + 1}/{NUM_ITERATIONS}"
        f" | Best fitness: {global_best_score:.6f}"
    )

    # Update velocities and positions

    r1 = rng.random(
        (NUM_PARTICLES, DIMENSIONS)
    )

    r2 = rng.random(
        (NUM_PARTICLES, DIMENSIONS)
    )

    cognitive_component = (
        COGNITIVE_COEFFICIENT
        * r1
        * (
            personal_best_positions
            - positions
        )
    )

    social_component = (
        SOCIAL_COEFFICIENT
        * r2
        * (
            global_best_position
            - positions
        )
    )

    velocities = (
        INERTIA_WEIGHT * velocities
        + cognitive_component
        + social_component
    )

    # Limit velocity to prevent
    # particles from moving too far.

    velocities = np.clip(
        velocities,
        -velocity_limits,
        velocity_limits
    )

    positions = positions + velocities

    # Keep particles inside bounds

    positions = np.clip(
        positions,
        LOWER_BOUNDS,
        UPPER_BOUNDS
    )

# Evaluate final positions as well.

for i in range(NUM_PARTICLES):

    score = pso_fitness(positions[i])

    if score < personal_best_scores[i]:

        personal_best_scores[i] = score

        personal_best_positions[i] = (
            positions[i].copy()
        )

best_index = int(
    np.argmin(personal_best_scores)
)

if personal_best_scores[best_index] < global_best_score:

    global_best_score = float(
        personal_best_scores[best_index]
    )

    global_best_position = (
        personal_best_positions[best_index].copy()
    )

best_params = decode_particle(
    global_best_position
)

print("\nPSO optimization completed!")

print("\nBest parameters:")
print(best_params)

print("\nBest validation error:")
print(global_best_score)

# ============================================================
# STEP 15: SAVE OPTIMIZATION HISTORY
# ============================================================

history_df = pd.DataFrame(
    optimization_history
)

history_df.to_csv(
    ROOT / "pso_optimization_history.csv",
    index=False
)

plt.figure(figsize=(12, 6))

plt.plot(
    history_df["Iteration"],
    history_df["Global_Best_Fitness"],
    marker="o",
    linewidth=2,
    label="Global Best Fitness"
)

plt.plot(
    history_df["Iteration"],
    history_df["Mean_Particle_Fitness"],
    marker="s",
    linestyle="--",
    label="Mean Particle Fitness"
)

plt.title(
    "PSO Optimization Convergence",
    fontsize=15,
    fontweight="bold"
)

plt.xlabel("Iteration")

plt.ylabel(
    "Normalized Validation MAE (Lower is Better)"
)

plt.legend()
plt.grid(alpha=0.3)

save_plot("pso_optimization_graph.png")

# ============================================================
# STEP 16: DEFINE BASELINE AND PSO MODELS
# ============================================================

models = {
    "Linear Regression": LinearRegression(),

    "Ridge Regression": Ridge(
        alpha=1.0
    ),

    "Decision Tree": DecisionTreeRegressor(
        max_depth=3,
        random_state=RANDOM_STATE
    ),

    "Random Forest": RandomForestRegressor(
        n_estimators=100,
        max_depth=4,
        random_state=RANDOM_STATE,
        n_jobs=-1
    ),

    "Gradient Boosting": MultiOutputRegressor(
        GradientBoostingRegressor(
            n_estimators=100,
            learning_rate=0.05,
            max_depth=2,
            random_state=RANDOM_STATE
        )
    ),

    "PSO Random Forest": make_pso_estimator(
        best_params
    )
}

# ============================================================
# STEP 17: TRAIN AND EVALUATE MODELS
# ============================================================

model_results = []
trained_models = {}

for name, estimator in models.items():

    print("\nTraining:", name)

    model = make_pipeline(estimator)

    model.fit(
        X_train,
        y_train
    )

    predictions = model.predict(X_test)

    predictions = np.maximum(
        predictions, 0
    )

    metrics = evaluate_predictions(
        y_test,
        predictions
    )

    error = normalized_mae(
        y_test,
        predictions,
        target_scales
    )

    model_results.append({
        "Model": name,
        **metrics,
        "Normalized_Error": error
    })

    trained_models[name] = model

    print(
        "Passenger R2:",
        round(metrics["Passenger_R2"], 4)
    )

    print(
        "Freight R2:",
        round(metrics["Freight_R2"], 4)
    )

# ============================================================
# STEP 18: SAVE MODEL COMPARISON CSV
# ============================================================

comparison_df = pd.DataFrame(
    model_results
)

comparison_df = comparison_df.sort_values(
    "Normalized_Error"
).reset_index(drop=True)

comparison_df.to_csv(
    ROOT / "pso_model_comparison.csv",
    index=False
)

print("\nModel comparison:")
print(comparison_df.to_string(index=False))

# ============================================================
# STEP 19: PSO ACCURACY GRAPH
# ============================================================

# R2 is a regression metric, not a percentage.
# Negative R2 scores are possible.

accuracy_df = comparison_df.set_index(
    "Model"
)[
    [
        "Passenger_R2",
        "Freight_R2"
    ]
]

accuracy_df.plot(
    kind="bar",
    figsize=(13, 7),
    width=0.75
)

plt.title(
    "PSO - Model Accuracy Comparison (R2 Score)",
    fontsize=15,
    fontweight="bold"
)

plt.xlabel("Machine Learning Model")
plt.ylabel("R2 Score")

plt.xticks(
    rotation=30,
    ha="right"
)

plt.axhline(
    y=0,
    color="black",
    linewidth=1
)

plt.legend([
    "Passenger R2",
    "Freight R2"
])

plt.grid(axis="y", alpha=0.3)

save_plot("pso_accuracy_graph.png")

# ============================================================
# STEP 20: PSO COMPARISON GRAPH
# ============================================================

plt.figure(figsize=(13, 7))

sns.barplot(
    data=comparison_df,
    x="Model",
    y="Normalized_Error",
    color="steelblue"
)

plt.title(
    "PSO Optimized vs Baseline Models",
    fontsize=15,
    fontweight="bold"
)

plt.xlabel("Machine Learning Model")

plt.ylabel(
    "Normalized Test MAE (Lower is Better)"
)

plt.xticks(
    rotation=30,
    ha="right"
)

plt.grid(axis="y", alpha=0.3)

save_plot("pso_comparison_graph.png")

# ============================================================
# STEP 21: SELECT PSO MODEL
# ============================================================

# Keep the PSO-optimized model as the
# final project model, regardless of
# which baseline has the lowest test error.

pso_model = trained_models[
    "PSO Random Forest"
]

pso_test_predictions = pso_model.predict(
    X_test
)

pso_test_predictions = np.maximum(
    pso_test_predictions, 0
)

pso_metrics = evaluate_predictions(
    y_test,
    pso_test_predictions
)

print("\nPSO model test performance:")

for metric, value in pso_metrics.items():

    print(
        metric,
        ":",
        round(value, 4)
    )

# ============================================================
# STEP 22: GENERATE RESULT CSV
# ============================================================

result_df = df.loc[
    test_mask,
    [
        "YEAR",
        "YEAR_START",
        "OPERATOR"
    ]
].copy()

result_df["Actual_Passengers"] = (
    y_test["PAX_TOTAL"].values
)

result_df["Predicted_Passengers"] = (
    pso_test_predictions[:, 0]
)

result_df["Actual_Freight"] = (
    y_test["FREIGHT_TOTAL"].values
)

result_df["Predicted_Freight"] = (
    pso_test_predictions[:, 1]
)

result_df["Passenger_Error"] = (
    result_df["Actual_Passengers"]
    - result_df["Predicted_Passengers"]
)

result_df["Freight_Error"] = (
    result_df["Actual_Freight"]
    - result_df["Predicted_Freight"]
)

result_df["Passenger_Absolute_Error"] = (
    result_df["Passenger_Error"].abs()
)

result_df["Freight_Absolute_Error"] = (
    result_df["Freight_Error"].abs()
)

result_df = result_df.sort_values(
    ["YEAR_START", "OPERATOR"]
).reset_index(drop=True)

result_df.to_csv(
    ROOT / "pso_result.csv",
    index=False
)

print("\nPSO result CSV:")
print(result_df.to_string(index=False))

# ============================================================
# STEP 23: GENERATE RESULT GRAPH
# ============================================================

plt.figure(figsize=(13, 7))

positions = np.arange(
    len(result_df)
)

labels = (
    result_df["YEAR"].astype(str)
    + " - "
    + result_df["OPERATOR"].astype(str)
)

plt.plot(
    positions,
    result_df["Actual_Passengers"],
    marker="o",
    linewidth=2,
    label="Actual Passengers"
)

plt.plot(
    positions,
    result_df["Predicted_Passengers"],
    marker="s",
    linestyle="--",
    linewidth=2,
    label="PSO Predicted Passengers"
)

plt.xticks(
    positions,
    labels,
    rotation=45,
    ha="right"
)

plt.title(
    "PSO - Actual vs Predicted Passenger Traffic",
    fontsize=15,
    fontweight="bold"
)

plt.xlabel("Financial Year and Operator")
plt.ylabel("Passenger Traffic")

plt.legend()
plt.grid(alpha=0.3)

save_plot("pso_result_graph.png")

# ============================================================
# STEP 24: REFIT PSO MODEL ON ALL HISTORICAL DATA
# ============================================================

# Test evaluation is already completed.
# Refit for future forecasting.

final_pso_model = make_pipeline(
    make_pso_estimator(best_params)
)

final_pso_model.fit(X, y)

# ============================================================
# STEP 25: SAVE PSO MODEL AS PKL
# ============================================================

pkl_path = ROOT / "pso_model.pkl"

joblib.dump(
    final_pso_model,
    pkl_path
)

print("\nSaved:", pkl_path)

# ============================================================
# STEP 26: SAVE PSO MODEL AS H5
# ============================================================

# H5 stores a pickled sklearn pipeline.
# This is NOT a native Keras H5 model.

h5_path = ROOT / "pso_model.h5"

model_bytes = pickle.dumps(
    final_pso_model,
    protocol=pickle.HIGHEST_PROTOCOL
)

model_array = np.frombuffer(
    model_bytes,
    dtype=np.uint8
)

with h5py.File(h5_path, "w") as h5:

    h5.create_dataset(
        "serialized_model",
        data=model_array,
        compression="gzip"
    )

    h5.attrs["algorithm"] = (
        "PSO Optimized Random Forest"
    )

    h5.attrs["format"] = (
        "Pickled scikit-learn pipeline"
    )

    h5.attrs["features"] = json.dumps(
        FEATURES
    )

    h5.attrs["targets"] = json.dumps(
        TARGETS
    )

print("Saved:", h5_path)

# ============================================================
# STEP 27: SAVE JSON AND YAML METADATA
# ============================================================

metadata = {
    "project": (
        "AI-Based International Air Traffic Analysis"
    ),
    "optimization": (
        "Particle Swarm Optimization"
    ),
    "model": "RandomForestRegressor",
    "features": FEATURES,
    "targets": TARGETS,
    "best_hyperparameters": best_params,
    "pso_settings": {
        "num_particles": NUM_PARTICLES,
        "num_iterations": NUM_ITERATIONS,
        "inertia_weight": INERTIA_WEIGHT,
        "cognitive_coefficient": (
            COGNITIVE_COEFFICIENT
        ),
        "social_coefficient": (
            SOCIAL_COEFFICIENT
        )
    },
    "best_validation_error": float(
        global_best_score
    ),
    "test_metrics": {
        key: float(value)
        for key, value in pso_metrics.items()
    },
    "training_years": [
        int(year) for year in train_years
    ],
    "testing_years": [
        int(year) for year in test_years
    ],
    "note": (
        "Exploratory forecasts based on a small "
        "historical dataset. H5 stores a serialized "
        "scikit-learn pipeline, not a Keras model."
    )
}

json_path = ROOT / "pso_model.json"

with open(
    json_path,
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        metadata,
        f,
        indent=4
    )

yaml_path = ROOT / "pso_model.yaml"

with open(
    yaml_path,
    "w",
    encoding="utf-8"
) as f:

    yaml.safe_dump(
        metadata,
        f,
        sort_keys=False
    )

print("Saved:", json_path)
print("Saved:", yaml_path)

# ============================================================
# STEP 28: GENERATE FUTURE PSO PREDICTIONS
# ============================================================

last_year = int(
    df["YEAR_START"].max()
)

future_years = list(
    range(last_year + 1, last_year + 6)
)

available_operators = set(
    df["OPERATOR"].unique()
)

if {
    "INDIAN",
    "FOREIGN"
}.issubset(available_operators):

    forecast_operators = [
        "INDIAN",
        "FOREIGN"
    ]

else:

    forecast_operators = sorted(
        op for op in available_operators
        if op != "TOTAL"
    )

    if not forecast_operators:
        forecast_operators = ["TOTAL"]

future_records = []

for year in future_years:

    for operator in forecast_operators:

        future_records.append({
            "YEAR_START": year,
            "OPERATOR": operator
        })

future_df = pd.DataFrame(
    future_records
)

future_predictions = final_pso_model.predict(
    future_df[FEATURES]
)

future_predictions = np.maximum(
    future_predictions, 0
)

future_df["Predicted_Passengers"] = (
    future_predictions[:, 0]
)

future_df["Predicted_Freight"] = (
    future_predictions[:, 1]
)

# Compute TOTAL from component operators
# to avoid double-counting.

if (
    "INDIAN" in forecast_operators
    and "FOREIGN" in forecast_operators
):

    total_df = future_df.groupby(
        "YEAR_START",
        as_index=False
    )[
        [
            "Predicted_Passengers",
            "Predicted_Freight"
        ]
    ].sum()

    total_df["OPERATOR"] = "TOTAL"

    future_df = pd.concat(
        [future_df, total_df],
        ignore_index=True
    )

future_df["YEAR"] = future_df[
    "YEAR_START"
].apply(
    lambda year: (
        f"{year}-{str((year + 1) % 100).zfill(2)}"
    )
)

future_df = future_df[
    [
        "YEAR",
        "YEAR_START",
        "OPERATOR",
        "Predicted_Passengers",
        "Predicted_Freight"
    ]
]

future_df = future_df.sort_values(
    ["YEAR_START", "OPERATOR"]
).reset_index(drop=True)

future_df.to_csv(
    ROOT / "pso_prediction.csv",
    index=False
)

print("\nPSO future predictions:")
print(future_df.to_string(index=False))

# ============================================================
# STEP 29: PREPARE HISTORICAL TOTAL TRAFFIC
# ============================================================

historical_total = df[
    df["OPERATOR"] == "TOTAL"
].copy()

if historical_total.empty:

    historical_total = df.groupby(
        "YEAR_START",
        as_index=False
    )[
        [
            "PAX_TOTAL",
            "FREIGHT_TOTAL"
        ]
    ].sum()

future_total = future_df[
    future_df["OPERATOR"] == "TOTAL"
].copy()

if future_total.empty:

    future_total = future_df.groupby(
        "YEAR_START",
        as_index=False
    )[
        [
            "Predicted_Passengers",
            "Predicted_Freight"
        ]
    ].sum()

# ============================================================
# STEP 30: PASSENGER PREDICTION GRAPH
# ============================================================

plt.figure(figsize=(13, 7))

plt.plot(
    historical_total["YEAR_START"],
    historical_total["PAX_TOTAL"],
    marker="o",
    linewidth=2,
    label="Historical Passenger Traffic"
)

plt.plot(
    future_total["YEAR_START"],
    future_total["Predicted_Passengers"],
    marker="s",
    linestyle="--",
    linewidth=2,
    label="PSO Passenger Forecast"
)

plt.axvline(
    last_year,
    color="gray",
    linestyle=":",
    label="Forecast Boundary"
)

plt.title(
    "PSO - International Passenger Traffic Forecast",
    fontsize=15,
    fontweight="bold"
)

plt.xlabel("Financial Year Starting")
plt.ylabel("Passenger Traffic")

plt.legend()
plt.grid(alpha=0.3)

save_plot("pso_prediction_graph.png")

# ============================================================
# STEP 31: FREIGHT PREDICTION GRAPH
# ============================================================

plt.figure(figsize=(13, 7))

plt.plot(
    historical_total["YEAR_START"],
    historical_total["FREIGHT_TOTAL"],
    marker="o",
    linewidth=2,
    label="Historical Freight Traffic"
)

plt.plot(
    future_total["YEAR_START"],
    future_total["Predicted_Freight"],
    marker="s",
    linestyle="--",
    linewidth=2,
    label="PSO Freight Forecast"
)

plt.axvline(
    last_year,
    color="gray",
    linestyle=":",
    label="Forecast Boundary"
)

plt.title(
    "PSO - International Freight Traffic Forecast",
    fontsize=15,
    fontweight="bold"
)

plt.xlabel("Financial Year Starting")
plt.ylabel("Freight Traffic")

plt.legend()
plt.grid(alpha=0.3)

save_plot("pso_freight_prediction_graph.png")

# ============================================================
# STEP 32: PASSENGER TRAFFIC TREND GRAPH
# ============================================================

plt.figure(figsize=(13, 7))

for operator in sorted(
    df["OPERATOR"].unique()
):

    subset = df[
        df["OPERATOR"] == operator
    ].sort_values("YEAR_START")

    plt.plot(
        subset["YEAR_START"],
        subset["PAX_TOTAL"],
        marker="o",
        linewidth=2,
        label=operator
    )

plt.title(
    "PSO - Historical Passenger Traffic Trends",
    fontsize=15,
    fontweight="bold"
)

plt.xlabel("Financial Year Starting")
plt.ylabel("Passenger Traffic")

plt.legend()
plt.grid(alpha=0.3)

save_plot("pso_passenger_trend_graph.png")

# ============================================================
# STEP 33: FREIGHT TRAFFIC TREND GRAPH
# ============================================================

plt.figure(figsize=(13, 7))

for operator in sorted(
    df["OPERATOR"].unique()
):

    subset = df[
        df["OPERATOR"] == operator
    ].sort_values("YEAR_START")

    plt.plot(
        subset["YEAR_START"],
        subset["FREIGHT_TOTAL"],
        marker="o",
        linewidth=2,
        label=operator
    )

plt.title(
    "PSO - Historical Freight Traffic Trends",
    fontsize=15,
    fontweight="bold"
)

plt.xlabel("Financial Year Starting")
plt.ylabel("Freight Traffic")

plt.legend()
plt.grid(alpha=0.3)

save_plot("pso_freight_trend_graph.png")

# ============================================================
# STEP 34: SAVE MODEL SUMMARY CSV
# ============================================================

summary = {
    "Algorithm": "PSO Optimized Random Forest",
    "PSO_Best_Validation_Error": (
        global_best_score
    ),
    "Training_Samples": len(X_train),
    "Testing_Samples": len(X_test),
    **pso_metrics
}

summary_df = pd.DataFrame([
    summary
])

summary_df.to_csv(
    ROOT / "pso_model_summary.csv",
    index=False
)

# ============================================================
# STEP 35: VERIFY SAVED MODEL
# ============================================================

loaded_model = joblib.load(
    ROOT / "pso_model.pkl"
)

sample_input = X.tail(3).copy()

sample_predictions = loaded_model.predict(
    sample_input
)

print("\nSaved PKL model verification:")
print(sample_predictions)

# Verify H5 serialization.
# Never unpickle files from untrusted sources.

with h5py.File(
    ROOT / "pso_model.h5",
    "r"
) as h5:

    model_data = (
        h5["serialized_model"][:].tobytes()
    )

restored_model = pickle.loads(
    model_data
)

restored_predictions = restored_model.predict(
    sample_input
)

assert np.allclose(
    sample_predictions,
    restored_predictions
)

print("\nH5 verification successful!")

# ============================================================
# STEP 36: FINAL OUTPUT SUMMARY
# ============================================================

print("\n" + "=" * 70)
print("PSO PROJECT COMPLETED SUCCESSFULLY")
print("=" * 70)

print("\nBest PSO hyperparameters:")
print(best_params)

print("\nPSO Test Metrics:")

for key, value in pso_metrics.items():

    print(
        f"{key}: {value:.4f}"
    )

print("\nGenerated PSO files:")

generated_files = sorted(
    ROOT.glob("pso_*")
)

for file in generated_files:

    print("[SAVED]", file.name)

print("\nTotal PSO files:", len(generated_files))

print("\nAll files saved in:")
print(ROOT)

print("\nIMPORTANT:")
print(
    "The historical dataset is small. "
    "PSO optimization and forecasts are exploratory."
)
```
