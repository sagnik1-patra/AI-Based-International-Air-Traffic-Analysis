```python
import os
import json
import pickle
import warnings
from pathlib import Path

import h5py
import joblib
import yaml
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
from sklearn.impute import SimpleImputer
from sklearn.multioutput import MultiOutputRegressor
from sklearn.linear_model import LinearRegression, Ridge
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import (
    RandomForestRegressor,
    GradientBoostingRegressor
)
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

warnings.filterwarnings("ignore")

# ============================================================
# STEP 1: SET PROJECT PATH
# ============================================================

PROJECT_DIR = Path(
    r"C:\Users\sagni\Downloads"
    r"\AI-Based International Air Traffic Analysis"
)

DATASET_PATH = PROJECT_DIR / "IntlTrafficData.csv"

PROJECT_DIR.mkdir(parents=True, exist_ok=True)

print("=" * 65)
print("AI-BASED INTERNATIONAL AIR TRAFFIC ANALYSIS")
print("=" * 65)

print("Project folder:", PROJECT_DIR)

# ============================================================
# STEP 2: LOAD DATASET
# ============================================================

if not DATASET_PATH.exists():
    raise FileNotFoundError(
        f"Dataset not found: {DATASET_PATH}"
    )

df = pd.read_csv(DATASET_PATH)

print("\nDataset loaded successfully!")
print("Dataset shape:", df.shape)

print("\nFirst five records:")
print(df.head())

print("\nDataset information:")
df.info()

# ============================================================
# STEP 3: CLEAN COLUMN NAMES
# ============================================================

df.columns = (
    df.columns
    .str.strip()
    .str.upper()
    .str.replace(r"\s+", "_", regex=True)
)

print("\nCleaned columns:")
print(df.columns.tolist())

# Expected columns:
# YEAR, OPERATOR, PAX_TO, PAX_FROM,
# PAX_TOTAL, FREIGHT_TO, FREIGHT_FROM,
# FREIGHT_TOTAL

required_columns = [
    "YEAR",
    "OPERATOR",
    "PAX_TOTAL",
    "FREIGHT_TOTAL"
]

missing = [
    c for c in required_columns
    if c not in df.columns
]

if missing:
    raise ValueError(
        f"Missing required columns: {missing}"
    )

# ============================================================
# STEP 4: DATA PREPROCESSING
# ============================================================

df["YEAR"] = df["YEAR"].astype(str).str.strip()

df["OPERATOR"] = (
    df["OPERATOR"]
    .astype(str)
    .str.strip()
    .str.upper()
)

# Convert financial year 2005-06 into 2005
df["YEAR_START"] = pd.to_numeric(
    df["YEAR"].str.extract(r"(\d{4})")[0],
    errors="coerce"
)

numeric_columns = [
    "PAX_TO",
    "PAX_FROM",
    "PAX_TOTAL",
    "FREIGHT_TO",
    "FREIGHT_FROM",
    "FREIGHT_TOTAL"
]

for col in numeric_columns:
    if col in df.columns:
        df[col] = pd.to_numeric(
            df[col].astype(str).str.replace(
                ",", "", regex=False
            ),
            errors="coerce"
        )

df = df.dropna(
    subset=[
        "YEAR_START",
        "OPERATOR",
        "PAX_TOTAL",
        "FREIGHT_TOTAL"
    ]
).copy()

df["YEAR_START"] = df["YEAR_START"].astype(int)

df = df.sort_values(
    ["YEAR_START", "OPERATOR"]
).reset_index(drop=True)

print("\nMissing values:")
print(df.isnull().sum())

print("\nStatistical summary:")
print(df.describe(include="all"))

# ============================================================
# STEP 5: BASIC TRAFFIC ANALYSIS
# ============================================================

print("\nPassenger traffic by operator:")

passenger_summary = df.groupby(
    "OPERATOR"
)["PAX_TOTAL"].agg(
    ["mean", "min", "max", "sum"]
)

print(passenger_summary)

print("\nFreight traffic by operator:")

freight_summary = df.groupby(
    "OPERATOR"
)["FREIGHT_TOTAL"].agg(
    ["mean", "min", "max", "sum"]
)

print(freight_summary)

# Save processed dataset
df.to_csv(
    PROJECT_DIR / "processed_air_traffic.csv",
    index=False
)

# ============================================================
# STEP 6: FEATURE SELECTION
# ============================================================

# We predict traffic using only year and operator.
# Actual passenger/freight totals are NOT used
# as predictors because that would leak target data.

feature_columns = [
    "YEAR_START",
    "OPERATOR"
]

target_columns = [
    "PAX_TOTAL",
    "FREIGHT_TOTAL"
]

X = df[feature_columns].copy()
y = df[target_columns].copy()

print("\nInput features:", feature_columns)
print("Prediction targets:", target_columns)

# ============================================================
# STEP 7: TIME-AWARE TRAIN/TEST SPLIT
# ============================================================

# Split by complete years to prevent future-year
# information from entering the training data.

years = sorted(df["YEAR_START"].unique())

if len(years) < 3:
    raise ValueError(
        "At least three different years are required."
    )

split_index = max(
    2,
    int(len(years) * 0.70)
)

split_index = min(
    split_index,
    len(years) - 1
)

train_years = years[:split_index]
test_years = years[split_index:]

train_mask = df["YEAR_START"].isin(train_years)
test_mask = df["YEAR_START"].isin(test_years)

X_train = X.loc[train_mask].copy()
X_test = X.loc[test_mask].copy()

y_train = y.loc[train_mask].copy()
y_test = y.loc[test_mask].copy()

print("\nTraining years:", train_years)
print("Testing years:", test_years)

print("Training samples:", len(X_train))
print("Testing samples:", len(X_test))

# ============================================================
# STEP 8: CREATE PREPROCESSOR
# ============================================================

def create_preprocessor():

    numeric_transformer = Pipeline([
        (
            "imputer",
            SimpleImputer(strategy="median")
        )
    ])

    categorical_transformer = Pipeline([
        (
            "imputer",
            SimpleImputer(strategy="most_frequent")
        ),
        (
            "encoder",
            OneHotEncoder(
                handle_unknown="ignore"
            )
        )
    ])

    preprocessor = ColumnTransformer([
        (
            "numeric",
            numeric_transformer,
            ["YEAR_START"]
        ),
        (
            "categorical",
            categorical_transformer,
            ["OPERATOR"]
        )
    ])

    return preprocessor

# ============================================================
# STEP 9: DEFINE MACHINE LEARNING MODELS
# ============================================================

models = {
    "Linear Regression": LinearRegression(),

    "Ridge Regression": Ridge(
        alpha=1.0
    ),

    "Decision Tree": DecisionTreeRegressor(
        max_depth=3,
        random_state=42
    ),

    "Random Forest": RandomForestRegressor(
        n_estimators=100,
        max_depth=4,
        random_state=42
    ),

    "Gradient Boosting": MultiOutputRegressor(
        GradientBoostingRegressor(
            n_estimators=100,
            learning_rate=0.05,
            max_depth=2,
            random_state=42
        )
    )
}

# ============================================================
# STEP 10: TRAIN AND EVALUATE MODELS
# ============================================================

results = []
trained_models = {}

for name, estimator in models.items():

    print("\nTraining:", name)

    pipeline = Pipeline([
        (
            "preprocessor",
            create_preprocessor()
        ),
        (
            "model",
            estimator
        )
    ])

    pipeline.fit(X_train, y_train)

    predictions = pipeline.predict(X_test)

    predictions = np.maximum(
        predictions, 0
    )

    # Passenger and freight volumes have different
    # scales, so evaluate them separately.

    pax_mae = mean_absolute_error(
        y_test["PAX_TOTAL"],
        predictions[:, 0]
    )

    freight_mae = mean_absolute_error(
        y_test["FREIGHT_TOTAL"],
        predictions[:, 1]
    )

    pax_rmse = np.sqrt(
        mean_squared_error(
            y_test["PAX_TOTAL"],
            predictions[:, 0]
        )
    )

    freight_rmse = np.sqrt(
        mean_squared_error(
            y_test["FREIGHT_TOTAL"],
            predictions[:, 1]
        )
    )

    pax_r2 = r2_score(
        y_test["PAX_TOTAL"],
        predictions[:, 0]
    )

    freight_r2 = r2_score(
        y_test["FREIGHT_TOTAL"],
        predictions[:, 1]
    )

    # Normalized errors allow comparison across
    # passenger and freight targets.

    pax_scale = max(
        float(y_train["PAX_TOTAL"].mean()),
        1.0
    )

    freight_scale = max(
        float(y_train["FREIGHT_TOTAL"].mean()),
        1.0
    )

    normalized_error = (
        pax_mae / pax_scale
        + freight_mae / freight_scale
    ) / 2

    results.append({
        "Model": name,
        "Passenger_MAE": pax_mae,
        "Freight_MAE": freight_mae,
        "Passenger_RMSE": pax_rmse,
        "Freight_RMSE": freight_rmse,
        "Passenger_R2": pax_r2,
        "Freight_R2": freight_r2,
        "Normalized_Error": normalized_error
    })

    trained_models[name] = pipeline

    print("Passenger MAE:", round(pax_mae, 2))
    print("Freight MAE:", round(freight_mae, 2))
    print("Passenger R2:", round(pax_r2, 4))
    print("Freight R2:", round(freight_r2, 4))

# ============================================================
# STEP 11: COMPARE MODELS
# ============================================================

comparison_df = pd.DataFrame(results)

comparison_df = comparison_df.sort_values(
    "Normalized_Error",
    ascending=True
).reset_index(drop=True)

print("\nMODEL COMPARISON")
print(comparison_df.to_string(index=False))

comparison_df.to_csv(
    PROJECT_DIR / "model_comparison.csv",
    index=False
)

best_model_name = comparison_df.iloc[0]["Model"]

best_model = trained_models[best_model_name]

print("\nBest model:", best_model_name)

# ============================================================
# STEP 12: GENERATE TEST PREDICTIONS
# ============================================================

test_predictions = best_model.predict(X_test)

test_predictions = np.maximum(
    test_predictions, 0
)

prediction_df = df.loc[
    test_mask,
    ["YEAR", "OPERATOR"]
].copy()

prediction_df["Actual_Passengers"] = (
    y_test["PAX_TOTAL"].values
)

prediction_df["Predicted_Passengers"] = (
    test_predictions[:, 0]
)

prediction_df["Actual_Freight"] = (
    y_test["FREIGHT_TOTAL"].values
)

prediction_df["Predicted_Freight"] = (
    test_predictions[:, 1]
)

prediction_df["Passenger_Error"] = (
    prediction_df["Actual_Passengers"]
    - prediction_df["Predicted_Passengers"]
)

prediction_df["Freight_Error"] = (
    prediction_df["Actual_Freight"]
    - prediction_df["Predicted_Freight"]
)

prediction_df.to_csv(
    PROJECT_DIR / "predictions.csv",
    index=False
)

print("\nTest predictions:")
print(prediction_df.to_string(index=False))

# ============================================================
# STEP 13: REFIT SELECTED MODEL ON ALL AVAILABLE DATA
# ============================================================

# The holdout evaluation above is preserved.
# Refit for the final deployment artifact.

final_model = Pipeline([
    (
        "preprocessor",
        create_preprocessor()
    ),
    (
        "model",
        models[best_model_name]
    )
])

final_model.fit(X, y)

# ============================================================
# STEP 14: SAVE MODEL AS PKL
# ============================================================

pkl_path = PROJECT_DIR / "air_traffic_model.pkl"

joblib.dump(
    final_model,
    pkl_path
)

print("\nPKL model saved:", pkl_path)

# ============================================================
# STEP 15: SAVE MODEL AS H5
# ============================================================

# Store the serialized sklearn pipeline in HDF5.
# This is NOT a native Keras model.
# Only deserialize files from trusted sources.

h5_path = PROJECT_DIR / "air_traffic_model.h5"

model_bytes = pickle.dumps(
    final_model,
    protocol=pickle.HIGHEST_PROTOCOL
)

model_array = np.frombuffer(
    model_bytes,
    dtype=np.uint8
)

with h5py.File(h5_path, "w") as h5:

    h5.create_dataset(
        "serialized_model",
        data=model_array,
        compression="gzip"
    )

    h5.attrs["model_name"] = best_model_name

    h5.attrs["model_format"] = (
        "Pickled scikit-learn pipeline"
    )

    h5.attrs["features"] = json.dumps(
        feature_columns
    )

    h5.attrs["targets"] = json.dumps(
        target_columns
    )

print("H5 model saved:", h5_path)

# ============================================================
# STEP 16: SAVE MODEL METADATA AS JSON
# ============================================================

best_metrics = comparison_df.iloc[0]

metadata = {
    "project_name": (
        "AI-Based International Air Traffic Analysis"
    ),
    "dataset": DATASET_PATH.name,
    "model_name": best_model_name,
    "task": "Multi-output Regression",
    "features": feature_columns,
    "targets": target_columns,
    "training_samples": int(len(X_train)),
    "testing_samples": int(len(X_test)),
    "training_years": [
        int(v) for v in train_years
    ],
    "testing_years": [
        int(v) for v in test_years
    ],
    "evaluation": {
        "passenger_mae": float(
            best_metrics["Passenger_MAE"]
        ),
        "freight_mae": float(
            best_metrics["Freight_MAE"]
        ),
        "passenger_rmse": float(
            best_metrics["Passenger_RMSE"]
        ),
        "freight_rmse": float(
            best_metrics["Freight_RMSE"]
        ),
        "passenger_r2": float(
            best_metrics["Passenger_R2"]
        ),
        "freight_r2": float(
            best_metrics["Freight_R2"]
        ),
        "normalized_error": float(
            best_metrics["Normalized_Error"]
        )
    },
    "model_files": {
        "pkl": "air_traffic_model.pkl",
        "h5": "air_traffic_model.h5"
    },
    "note": (
        "Small historical dataset. Forecasts are "
        "exploratory and not validated for long-term use."
    )
}

json_path = PROJECT_DIR / "air_traffic_model.json"

with open(json_path, "w", encoding="utf-8") as f:
    json.dump(
        metadata,
        f,
        indent=4
    )

print("JSON metadata saved:", json_path)

# ============================================================
# STEP 17: SAVE MODEL CONFIGURATION AS YAML
# ============================================================

yaml_path = PROJECT_DIR / "air_traffic_model.yaml"

with open(yaml_path, "w", encoding="utf-8") as f:
    yaml.safe_dump(
        metadata,
        f,
        sort_keys=False,
        allow_unicode=True
    )

print("YAML configuration saved:", yaml_path)

# ============================================================
# STEP 18: VERIFY SAVED PKL MODEL
# ============================================================

loaded_model = joblib.load(pkl_path)

sample_input = X.tail(3).copy()

sample_predictions = loaded_model.predict(
    sample_input
)

print("\nPKL model verification:")
print(sample_input)

print("\nPredicted passenger and freight totals:")
print(sample_predictions)

# ============================================================
# STEP 19: VERIFY SAVED H5 MODEL
# ============================================================

with h5py.File(h5_path, "r") as h5:
    restored_bytes = (
        h5["serialized_model"][:].tobytes()
    )

# Safe here because this file was created locally
# by this script. Never unpickle untrusted files.

restored_model = pickle.loads(
    restored_bytes
)

h5_predictions = restored_model.predict(
    sample_input
)

assert np.allclose(
    sample_predictions,
    h5_predictions
)

print("\nH5 model verification successful!")

# ============================================================
# STEP 20: DISPLAY GENERATED FILES
# ============================================================

print("\n" + "=" * 65)
print("PROJECT EXECUTION COMPLETED")
print("=" * 65)

print("\nBest model:", best_model_name)

print("\nGenerated files:")

generated_files = [
    "processed_air_traffic.csv",
    "model_comparison.csv",
    "predictions.csv",
    "air_traffic_model.pkl",
    "air_traffic_model.h5",
    "air_traffic_model.yaml",
    "air_traffic_model.json"
]

for filename in generated_files:
    file_path = PROJECT_DIR / filename

    if file_path.exists():
        print("[SAVED]", filename)
    else:
        print("[MISSING]", filename)

print("\nAll outputs stored in:")
print(PROJECT_DIR)
```
