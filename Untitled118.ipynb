```python
import json
import pickle
import warnings
from pathlib import Path

import h5py
import joblib
import yaml
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
from sklearn.impute import SimpleImputer
from sklearn.linear_model import LinearRegression, Ridge
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import (
    RandomForestRegressor,
    GradientBoostingRegressor
)
from sklearn.multioutput import MultiOutputRegressor
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

warnings.filterwarnings("ignore")

# ============================================================
# 1. PROJECT CONFIGURATION
# ============================================================

ROOT = Path(
    r"C:\Users\sagni\Downloads"
    r"\AI-Based International Air Traffic Analysis"
)

DATA_PATH = ROOT / "IntlTrafficData.csv"

ROOT.mkdir(parents=True, exist_ok=True)

RANDOM_STATE = 42
rng = np.random.default_rng(RANDOM_STATE)

sns.set_theme(style="whitegrid")

plt.rcParams.update({
    "figure.figsize": (12, 6),
    "font.size": 11,
    "savefig.dpi": 300
})

print("=" * 70)
print("AIS INTERNATIONAL AIR TRAFFIC ANALYSIS")
print("=" * 70)

# ============================================================
# 2. SAVE AND DISPLAY GRAPHS
# ============================================================

def save_plot(filename):

    path = ROOT / filename

    plt.tight_layout()

    plt.savefig(
        path,
        dpi=300,
        bbox_inches="tight"
    )

    print("[SAVED]", filename)

    plt.show()
    plt.close()

# ============================================================
# 3. LOAD DATASET
# ============================================================

if not DATA_PATH.exists():
    raise FileNotFoundError(
        f"Dataset not found: {DATA_PATH}"
    )

df = pd.read_csv(DATA_PATH)

print("\nOriginal dataset shape:", df.shape)
print(df.head())

# ============================================================
# 4. DATA CLEANING
# ============================================================

df.columns = (
    df.columns
    .str.strip()
    .str.upper()
    .str.replace(r"\s+", "_", regex=True)
)

required = [
    "YEAR",
    "OPERATOR",
    "PAX_TOTAL",
    "FREIGHT_TOTAL"
]

missing = [
    col for col in required
    if col not in df.columns
]

if missing:
    raise ValueError(
        f"Missing required columns: {missing}"
    )

df["YEAR"] = df["YEAR"].astype(str).str.strip()

df["OPERATOR"] = (
    df["OPERATOR"]
    .astype(str)
    .str.strip()
    .str.upper()
)

df["YEAR_START"] = pd.to_numeric(
    df["YEAR"].str.extract(r"(\d{4})")[0],
    errors="coerce"
)

numeric_cols = [
    "PAX_TO",
    "PAX_FROM",
    "PAX_TOTAL",
    "FREIGHT_TO",
    "FREIGHT_FROM",
    "FREIGHT_TOTAL"
]

for col in numeric_cols:

    if col in df.columns:

        df[col] = pd.to_numeric(
            df[col].astype(str).str.replace(
                ",", "", regex=False
            ),
            errors="coerce"
        )

df = df.dropna(
    subset=[
        "YEAR_START",
        "OPERATOR",
        "PAX_TOTAL",
        "FREIGHT_TOTAL"
    ]
).copy()

df["YEAR_START"] = df["YEAR_START"].astype(int)

df = df.sort_values(
    ["YEAR_START", "OPERATOR"]
).reset_index(drop=True)

df.to_csv(
    ROOT / "ais_processed_data.csv",
    index=False
)

print("\nCleaned dataset:")
print(df.head())

# ============================================================
# 5. CORRELATION HEATMAP
# ============================================================

heat_cols = [
    col for col in numeric_cols
    if col in df.columns
]

heat_data = df[
    heat_cols + ["YEAR_START"]
].select_dtypes(include=[np.number])

plt.figure(figsize=(12, 9))

sns.heatmap(
    heat_data.corr(),
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    linewidths=0.5,
    square=True
)

plt.title(
    "AIS - International Air Traffic Correlation Heatmap",
    fontsize=15,
    fontweight="bold"
)

save_plot("ais_heatmap.png")

# ============================================================
# 6. FEATURES AND TARGETS
# ============================================================

FEATURES = [
    "YEAR_START",
    "OPERATOR"
]

TARGETS = [
    "PAX_TOTAL",
    "FREIGHT_TOTAL"
]

# Avoid using PAX_TO / PAX_FROM as predictors
# of PAX_TOTAL, which would cause target leakage.

X = df[FEATURES].copy()
y = df[TARGETS].copy()

# ============================================================
# 7. TIME-AWARE HOLDOUT SPLIT
# ============================================================

years = sorted(df["YEAR_START"].unique())

if len(years) < 4:
    raise ValueError(
        "At least four distinct years are required "
        "for this AIS training and evaluation setup."
    )

split_idx = max(
    3,
    int(len(years) * 0.70)
)

split_idx = min(
    split_idx,
    len(years) - 1
)

train_years = years[:split_idx]
test_years = years[split_idx:]

train_mask = df["YEAR_START"].isin(train_years)
test_mask = df["YEAR_START"].isin(test_years)

X_train = X.loc[train_mask].copy()
X_test = X.loc[test_mask].copy()

y_train = y.loc[train_mask].copy()
y_test = y.loc[test_mask].copy()

print("\nTraining years:", train_years)
print("Testing years:", test_years)

# ============================================================
# 8. PREPROCESSOR
# ============================================================

def make_preprocessor():

    numeric_pipeline = Pipeline([
        (
            "imputer",
            SimpleImputer(strategy="median")
        )
    ])

    categorical_pipeline = Pipeline([
        (
            "imputer",
            SimpleImputer(strategy="most_frequent")
        ),
        (
            "encoder",
            OneHotEncoder(handle_unknown="ignore")
        )
    ])

    return ColumnTransformer([
        (
            "numeric",
            numeric_pipeline,
            ["YEAR_START"]
        ),
        (
            "categorical",
            categorical_pipeline,
            ["OPERATOR"]
        )
    ])

def make_pipeline(estimator):

    return Pipeline([
        (
            "preprocessor",
            make_preprocessor()
        ),
        (
            "model",
            estimator
        )
    ])

# ============================================================
# 9. EVALUATION FUNCTIONS
# ============================================================

def evaluate_predictions(y_true, y_pred):

    actual = np.asarray(y_true, dtype=float)
    predicted = np.maximum(
        np.asarray(y_pred, dtype=float),
        0
    )

    passenger_mae = mean_absolute_error(
        actual[:, 0],
        predicted[:, 0]
    )

    freight_mae = mean_absolute_error(
        actual[:, 1],
        predicted[:, 1]
    )

    passenger_rmse = np.sqrt(
        mean_squared_error(
            actual[:, 0],
            predicted[:, 0]
        )
    )

    freight_rmse = np.sqrt(
        mean_squared_error(
            actual[:, 1],
            predicted[:, 1]
        )
    )

    passenger_r2 = r2_score(
        actual[:, 0],
        predicted[:, 0]
    )

    freight_r2 = r2_score(
        actual[:, 1],
        predicted[:, 1]
    )

    return {
        "Passenger_MAE": passenger_mae,
        "Freight_MAE": freight_mae,
        "Passenger_RMSE": passenger_rmse,
        "Freight_RMSE": freight_rmse,
        "Passenger_R2": passenger_r2,
        "Freight_R2": freight_r2
    }

def normalized_mae(
    y_true,
    y_pred,
    scales
):

    actual = np.asarray(y_true, dtype=float)
    predicted = np.maximum(
        np.asarray(y_pred, dtype=float),
        0
    )

    errors = np.mean(
        np.abs(actual - predicted),
        axis=0
    )

    return float(
        np.mean(errors / scales)
    )

# ============================================================
# 10. AIS HYPERPARAMETER SEARCH SPACE
# ============================================================

# Each antibody encodes:
# [n_estimators, max_depth,
#  min_samples_split, min_samples_leaf,
#  max_features]

LOWER = np.array([
    30, 2, 2, 1, 0.4
], dtype=float)

UPPER = np.array([
    150, 8, 8, 4, 1.0
], dtype=float)

def decode_antibody(antibody):

    a = np.clip(
        np.asarray(antibody, dtype=float),
        LOWER,
        UPPER
    )

    return {
        "n_estimators": int(round(a[0])),
        "max_depth": int(round(a[1])),
        "min_samples_split": int(round(a[2])),
        "min_samples_leaf": int(round(a[3])),
        "max_features": float(a[4])
    }

def make_ais_estimator(params):

    return RandomForestRegressor(
        **params,
        random_state=RANDOM_STATE,
        n_jobs=-1
    )

# ============================================================
# 11. EXPANDING-WINDOW VALIDATION
# ============================================================

# Validation is performed ONLY within the
# training years. The final test years are
# never used for AIS hyperparameter selection.

def build_validation_folds():

    unique_years = sorted(
        X_train["YEAR_START"].unique()
    )

    folds = []

    for i in range(2, len(unique_years)):

        fit_years = unique_years[:i]
        val_year = unique_years[i]

        fit_mask = X_train[
            "YEAR_START"
        ].isin(fit_years)

        val_mask = (
            X_train["YEAR_START"] == val_year
        )

        folds.append((
            X_train.loc[fit_mask],
            y_train.loc[fit_mask],
            X_train.loc[val_mask],
            y_train.loc[val_mask]
        ))

    if not folds:
        raise ValueError(
            "Insufficient training years for "
            "expanding-window validation."
        )

    return folds

validation_folds = build_validation_folds()

# Use a fixed scale derived from training data.
# This makes passenger and freight errors comparable.

target_scales = np.maximum(
    np.abs(
        y_train.mean(axis=0).to_numpy(
            dtype=float
        )
    ),
    1.0
)

fitness_cache = {}

# ============================================================
# 12. AIS FITNESS FUNCTION
# ============================================================

def ais_fitness(antibody):

    params = decode_antibody(antibody)

    key = tuple(params.items())

    if key in fitness_cache:
        return fitness_cache[key]

    fold_errors = []

    for (
        X_fit,
        y_fit,
        X_val,
        y_val
    ) in validation_folds:

        model = make_pipeline(
            make_ais_estimator(params)
        )

        model.fit(X_fit, y_fit)

        pred = model.predict(X_val)

        error = normalized_mae(
            y_val,
            pred,
            target_scales
        )

        fold_errors.append(error)

    score = float(np.mean(fold_errors))

    fitness_cache[key] = score

    return score

# ============================================================
# 13. AIS CLONAL SELECTION OPTIMIZATION
# ============================================================

POPULATION_SIZE = 8
GENERATIONS = 8
ELITE_COUNT = 3
CLONES_PER_ELITE = 3
RANDOM_REPLACEMENTS = 2

def random_antibody():

    return rng.uniform(
        LOWER,
        UPPER
    )

def mutate_antibody(
    antibody,
    mutation_strength
):

    mutation = rng.normal(
        loc=0.0,
        scale=mutation_strength,
        size=len(LOWER)
    )

    mutated = (
        antibody
        + mutation * (UPPER - LOWER)
    )

    return np.clip(
        mutated,
        LOWER,
        UPPER
    )

population = [
    random_antibody()
    for _ in range(POPULATION_SIZE)
]

best_antibody = None
best_fitness = float("inf")

optimization_history = []

print("\nStarting AIS optimization...")

for generation in range(GENERATIONS):

    # Evaluate antibody population

    scored = [
        (
            ais_fitness(antibody),
            antibody.copy()
        )
        for antibody in population
    ]

    scored.sort(
        key=lambda item: item[0]
    )

    current_score, current_best = scored[0]

    if current_score < best_fitness:

        best_fitness = current_score
        best_antibody = current_best.copy()

    optimization_history.append({
        "Generation": generation + 1,
        "Best_Fitness": best_fitness,
        "Generation_Best_Fitness": current_score,
        "Mean_Fitness": float(
            np.mean([
                item[0] for item in scored
            ])
        )
    })

    print(
        f"Generation {generation + 1}/{GENERATIONS}"
        f" | Best error: {best_fitness:.6f}"
    )

    # Select high-affinity antibodies

    elites = [
        antibody
        for _, antibody in scored[:ELITE_COUNT]
    ]

    # Preserve the best solution

    candidates = [
        best_antibody.copy()
    ]

    # Clonal expansion and hypermutation

    for rank, antibody in enumerate(elites):

        # Better antibodies receive
        # smaller mutations.

        strength = (
            0.08 + 0.08 * rank
        ) * (
            1 - 0.5 * generation
            / max(GENERATIONS - 1, 1)
        )

        candidates.append(
            antibody.copy()
        )

        for _ in range(CLONES_PER_ELITE):

            clone = mutate_antibody(
                antibody,
                mutation_strength=strength
            )

            candidates.append(clone)

    # Introduce diversity

    for _ in range(RANDOM_REPLACEMENTS):
        candidates.append(
            random_antibody()
        )

    # Select the strongest antibodies

    candidate_scores = [
        (
            ais_fitness(antibody),
            antibody.copy()
        )
        for antibody in candidates
    ]

    candidate_scores.sort(
        key=lambda item: item[0]
    )

    population = [
        antibody
        for _, antibody in candidate_scores[
            :POPULATION_SIZE
        ]
    ]

best_params = decode_antibody(
    best_antibody
)

print("\nAIS optimization completed!")

print("\nBest hyperparameters:")
print(best_params)

print("\nBest validation error:")
print(best_fitness)

# ============================================================
# 14. SAVE OPTIMIZATION HISTORY
# ============================================================

history_df = pd.DataFrame(
    optimization_history
)

history_df.to_csv(
    ROOT / "ais_optimization_history.csv",
    index=False
)

plt.figure(figsize=(12, 6))

plt.plot(
    history_df["Generation"],
    history_df["Best_Fitness"],
    marker="o",
    linewidth=2,
    label="Best Fitness"
)

plt.plot(
    history_df["Generation"],
    history_df["Mean_Fitness"],
    marker="s",
    linestyle="--",
    label="Population Mean Fitness"
)

plt.title(
    "AIS Clonal Selection Optimization",
    fontsize=15,
    fontweight="bold"
)

plt.xlabel("Generation")
plt.ylabel(
    "Normalized Validation MAE (Lower is Better)"
)

plt.legend()
plt.grid(alpha=0.3)

save_plot("ais_optimization_graph.png")

# ============================================================
# 15. DEFINE BASELINE AND AIS MODELS
# ============================================================

models = {
    "Linear Regression": LinearRegression(),

    "Ridge Regression": Ridge(
        alpha=1.0
    ),

    "Decision Tree": DecisionTreeRegressor(
        max_depth=3,
        random_state=42
    ),

    "Random Forest": RandomForestRegressor(
        n_estimators=100,
        max_depth=4,
        random_state=42,
        n_jobs=-1
    ),

    "Gradient Boosting": MultiOutputRegressor(
        GradientBoostingRegressor(
            n_estimators=100,
            learning_rate=0.05,
            max_depth=2,
            random_state=42
        )
    ),

    "AIS Random Forest": make_ais_estimator(
        best_params
    )
}

# ============================================================
# 16. TRAIN AND EVALUATE MODELS
# ============================================================

model_results = []
trained_models = {}

for name, estimator in models.items():

    print("\nTraining:", name)

    model = make_pipeline(estimator)

    model.fit(
        X_train,
        y_train
    )

    predictions = model.predict(X_test)

    predictions = np.maximum(
        predictions, 0
    )

    metrics = evaluate_predictions(
        y_test,
        predictions
    )

    error = normalized_mae(
        y_test,
        predictions,
        target_scales
    )

    model_results.append({
        "Model": name,
        **metrics,
        "Normalized_Error": error
    })

    trained_models[name] = model

    print(
        "Passenger R2:",
        round(metrics["Passenger_R2"], 4)
    )

    print(
        "Freight R2:",
        round(metrics["Freight_R2"], 4)
    )

# ============================================================
# 17. SAVE MODEL COMPARISON CSV
# ============================================================

comparison_df = pd.DataFrame(
    model_results
)

comparison_df = comparison_df.sort_values(
    "Normalized_Error"
).reset_index(drop=True)

comparison_df.to_csv(
    ROOT / "ais_model_comparison.csv",
    index=False
)

print("\nModel comparison:")
print(comparison_df.to_string(index=False))

# ============================================================
# 18. AIS ACCURACY GRAPH
# ============================================================

# R2 is NOT classification accuracy.
# Negative R2 scores are valid.

accuracy_df = comparison_df.set_index(
    "Model"
)[
    [
        "Passenger_R2",
        "Freight_R2"
    ]
]

ax = accuracy_df.plot(
    kind="bar",
    figsize=(13, 7),
    width=0.75
)

plt.title(
    "AIS - Model Accuracy Comparison (R2 Score)",
    fontsize=15,
    fontweight="bold"
)

plt.xlabel("Machine Learning Model")
plt.ylabel("R2 Score")

plt.xticks(
    rotation=30,
    ha="right"
)

plt.axhline(
    0,
    color="black",
    linewidth=1
)

plt.legend([
    "Passenger R2",
    "Freight R2"
])

plt.grid(axis="y", alpha=0.3)

save_plot("ais_accuracy_graph.png")

# ============================================================
# 19. AIS COMPARISON GRAPH
# ============================================================

plt.figure(figsize=(13, 7))

sns.barplot(
    data=comparison_df,
    x="Model",
    y="Normalized_Error",
    color="steelblue"
)

plt.title(
    "AIS Optimized vs Baseline Models",
    fontsize=15,
    fontweight="bold"
)

plt.xlabel("Model")

plt.ylabel(
    "Normalized Test MAE (Lower is Better)"
)

plt.xticks(
    rotation=30,
    ha="right"
)

plt.grid(axis="y", alpha=0.3)

save_plot("ais_comparison_graph.png")

# ============================================================
# 20. SELECT AIS MODEL
# ============================================================

# Keep the AIS model as the final project model.
# Do not choose the best model using test scores.

ais_model = trained_models[
    "AIS Random Forest"
]

ais_test_predictions = ais_model.predict(
    X_test
)

ais_test_predictions = np.maximum(
    ais_test_predictions, 0
)

ais_metrics = evaluate_predictions(
    y_test,
    ais_test_predictions
)

print("\nAIS model performance:")

for metric, value in ais_metrics.items():
    print(metric, ":", round(value, 4))

# ============================================================
# 21. SAVE AIS RESULT CSV
# ============================================================

result_df = df.loc[
    test_mask,
    [
        "YEAR",
        "YEAR_START",
        "OPERATOR"
    ]
].copy()

result_df["Actual_Passengers"] = (
    y_test["PAX_TOTAL"].values
)

result_df["Predicted_Passengers"] = (
    ais_test_predictions[:, 0]
)

result_df["Actual_Freight"] = (
    y_test["FREIGHT_TOTAL"].values
)

result_df["Predicted_Freight"] = (
    ais_test_predictions[:, 1]
)

result_df["Passenger_Error"] = (
    result_df["Actual_Passengers"]
    - result_df["Predicted_Passengers"]
)

result_df["Freight_Error"] = (
    result_df["Actual_Freight"]
    - result_df["Predicted_Freight"]
)

result_df["Passenger_Absolute_Error"] = (
    result_df["Passenger_Error"].abs()
)

result_df["Freight_Absolute_Error"] = (
    result_df["Freight_Error"].abs()
)

result_df = result_df.sort_values(
    ["YEAR_START", "OPERATOR"]
).reset_index(drop=True)

result_df.to_csv(
    ROOT / "ais_result.csv",
    index=False
)

print("\nAIS result CSV:")
print(result_df.to_string(index=False))

# ============================================================
# 22. AIS RESULT GRAPH
# ============================================================

plt.figure(figsize=(13, 7))

positions = np.arange(
    len(result_df)
)

labels = (
    result_df["YEAR"].astype(str)
    + " - "
    + result_df["OPERATOR"].astype(str)
)

plt.plot(
    positions,
    result_df["Actual_Passengers"],
    marker="o",
    linewidth=2,
    label="Actual Passengers"
)

plt.plot(
    positions,
    result_df["Predicted_Passengers"],
    marker="s",
    linestyle="--",
    linewidth=2,
    label="AIS Predicted Passengers"
)

plt.xticks(
    positions,
    labels,
    rotation=45,
    ha="right"
)

plt.title(
    "AIS - Actual vs Predicted Passenger Traffic",
    fontsize=15,
    fontweight="bold"
)

plt.xlabel("Financial Year and Operator")
plt.ylabel("Passenger Traffic")

plt.legend()
plt.grid(alpha=0.3)

save_plot("ais_result_graph.png")

# ============================================================
# 23. REFIT AIS MODEL ON ALL HISTORICAL DATA
# ============================================================

# The held-out results remain unchanged.
# This final model is used for future forecasts.

final_ais_model = make_pipeline(
    make_ais_estimator(best_params)
)

final_ais_model.fit(X, y)

# ============================================================
# 24. SAVE AIS MODEL AS PKL
# ============================================================

pkl_path = ROOT / "ais_model.pkl"

joblib.dump(
    final_ais_model,
    pkl_path
)

print("\nSaved:", pkl_path)

# ============================================================
# 25. SAVE AIS MODEL AS H5
# ============================================================

# Stores a serialized sklearn model inside HDF5.
# This is NOT a native Keras .h5 model.

h5_path = ROOT / "ais_model.h5"

model_bytes = pickle.dumps(
    final_ais_model,
    protocol=pickle.HIGHEST_PROTOCOL
)

model_array = np.frombuffer(
    model_bytes,
    dtype=np.uint8
)

with h5py.File(h5_path, "w") as h5:

    h5.create_dataset(
        "serialized_model",
        data=model_array,
        compression="gzip"
    )

    h5.attrs["algorithm"] = (
        "AIS Optimized Random Forest"
    )

    h5.attrs["format"] = (
        "Pickled scikit-learn pipeline"
    )

    h5.attrs["features"] = json.dumps(
        FEATURES
    )

    h5.attrs["targets"] = json.dumps(
        TARGETS
    )

print("Saved:", h5_path)

# ============================================================
# 26. SAVE AIS JSON AND YAML
# ============================================================

metadata = {
    "project": (
        "AI-Based International Air Traffic Analysis"
    ),
    "optimization": (
        "Artificial Immune System - CLONALG"
    ),
    "model": "RandomForestRegressor",
    "features": FEATURES,
    "targets": TARGETS,
    "best_hyperparameters": best_params,
    "ais_settings": {
        "population_size": POPULATION_SIZE,
        "generations": GENERATIONS,
        "elite_count": ELITE_COUNT,
        "clones_per_elite": CLONES_PER_ELITE,
        "random_replacements": RANDOM_REPLACEMENTS
    },
    "best_validation_error": float(
        best_fitness
    ),
    "test_metrics": {
        key: float(value)
        for key, value in ais_metrics.items()
    },
    "training_years": [
        int(year) for year in train_years
    ],
    "testing_years": [
        int(year) for year in test_years
    ],
    "note": (
        "Exploratory forecasting with a small "
        "historical dataset. H5 stores a serialized "
        "sklearn model, not a native Keras model."
    )
}

json_path = ROOT / "ais_model.json"

with open(
    json_path,
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        metadata,
        f,
        indent=4
    )

yaml_path = ROOT / "ais_model.yaml"

with open(
    yaml_path,
    "w",
    encoding="utf-8"
) as f:

    yaml.safe_dump(
        metadata,
        f,
        sort_keys=False
    )

print("Saved:", json_path)
print("Saved:", yaml_path)

# ============================================================
# 27. GENERATE FUTURE AIS PREDICTIONS
# ============================================================

last_year = int(
    df["YEAR_START"].max()
)

future_years = list(
    range(last_year + 1, last_year + 6)
)

available_operators = set(
    df["OPERATOR"].unique()
)

if {
    "INDIAN",
    "FOREIGN"
}.issubset(available_operators):

    forecast_operators = [
        "INDIAN",
        "FOREIGN"
    ]

else:

    forecast_operators = sorted(
        op for op in available_operators
        if op != "TOTAL"
    )

    if not forecast_operators:
        forecast_operators = ["TOTAL"]

future_records = []

for year in future_years:

    for operator in forecast_operators:

        future_records.append({
            "YEAR_START": year,
            "OPERATOR": operator
        })

future_df = pd.DataFrame(
    future_records
)

future_predictions = final_ais_model.predict(
    future_df[FEATURES]
)

future_predictions = np.maximum(
    future_predictions, 0
)

future_df["Predicted_Passengers"] = (
    future_predictions[:, 0]
)

future_df["Predicted_Freight"] = (
    future_predictions[:, 1]
)

# Compute TOTAL from component operators.
# Do not double-count the TOTAL category.

if (
    "INDIAN" in forecast_operators
    and "FOREIGN" in forecast_operators
):

    total_df = future_df.groupby(
        "YEAR_START",
        as_index=False
    )[
        [
            "Predicted_Passengers",
            "Predicted_Freight"
        ]
    ].sum()

    total_df["OPERATOR"] = "TOTAL"

    future_df = pd.concat(
        [future_df, total_df],
        ignore_index=True
    )

future_df["YEAR"] = future_df[
    "YEAR_START"
].apply(
    lambda year: (
        f"{year}-{str((year + 1) % 100).zfill(2)}"
    )
)

future_df = future_df[
    [
        "YEAR",
        "YEAR_START",
        "OPERATOR",
        "Predicted_Passengers",
        "Predicted_Freight"
    ]
]

future_df = future_df.sort_values(
    ["YEAR_START", "OPERATOR"]
).reset_index(drop=True)

future_df.to_csv(
    ROOT / "ais_prediction.csv",
    index=False
)

print("\nAIS future predictions:")
print(future_df.to_string(index=False))

# ============================================================
# 28. HISTORICAL TOTAL TRAFFIC
# ============================================================

historical_total = df[
    df["OPERATOR"] == "TOTAL"
].copy()

if historical_total.empty:

    historical_total = df.groupby(
        "YEAR_START",
        as_index=False
    )[
        [
            "PAX_TOTAL",
            "FREIGHT_TOTAL"
        ]
    ].sum()

future_total = future_df[
    future_df["OPERATOR"] == "TOTAL"
].copy()

if future_total.empty:

    future_total = future_df.groupby(
        "YEAR_START",
        as_index=False
    )[
        [
            "Predicted_Passengers",
            "Predicted_Freight"
        ]
    ].sum()

# ============================================================
# 29. AIS PASSENGER PREDICTION GRAPH
# ============================================================

plt.figure(figsize=(13, 7))

plt.plot(
    historical_total["YEAR_START"],
    historical_total["PAX_TOTAL"],
    marker="o",
    linewidth=2,
    label="Historical Passenger Traffic"
)

plt.plot(
    future_total["YEAR_START"],
    future_total["Predicted_Passengers"],
    marker="s",
    linestyle="--",
    linewidth=2,
    label="AIS Passenger Forecast"
)

plt.axvline(
    last_year,
    color="gray",
    linestyle=":",
    label="Forecast Boundary"
)

plt.title(
    "AIS - International Passenger Traffic Forecast",
    fontsize=15,
    fontweight="bold"
)

plt.xlabel("Financial Year Starting")
plt.ylabel("Passenger Traffic")

plt.legend()
plt.grid(alpha=0.3)

save_plot("ais_prediction_graph.png")

# ============================================================
# 30. AIS FREIGHT PREDICTION GRAPH
# ============================================================

plt.figure(figsize=(13, 7))

plt.plot(
    historical_total["YEAR_START"],
    historical_total["FREIGHT_TOTAL"],
    marker="o",
    linewidth=2,
    label="Historical Freight Traffic"
)

plt.plot(
    future_total["YEAR_START"],
    future_total["Predicted_Freight"],
    marker="s",
    linestyle="--",
    linewidth=2,
    label="AIS Freight Forecast"
)

plt.axvline(
    last_year,
    color="gray",
    linestyle=":",
    label="Forecast Boundary"
)

plt.title(
    "AIS - International Freight Traffic Forecast",
    fontsize=15,
    fontweight="bold"
)

plt.xlabel("Financial Year Starting")
plt.ylabel("Freight Traffic")

plt.legend()
plt.grid(alpha=0.3)

save_plot("ais_freight_prediction_graph.png")

# ============================================================
# 31. AIS PASSENGER TREND GRAPH
# ============================================================

plt.figure(figsize=(13, 7))

for operator in sorted(
    df["OPERATOR"].unique()
):

    subset = df[
        df["OPERATOR"] == operator
    ].sort_values("YEAR_START")

    plt.plot(
        subset["YEAR_START"],
        subset["PAX_TOTAL"],
        marker="o",
        linewidth=2,
        label=operator
    )

plt.title(
    "AIS - Historical Passenger Traffic Trends",
    fontsize=15,
    fontweight="bold"
)

plt.xlabel("Financial Year Starting")
plt.ylabel("Passenger Traffic")

plt.legend()
plt.grid(alpha=0.3)

save_plot("ais_passenger_trend_graph.png")

# ============================================================
# 32. AIS FREIGHT TREND GRAPH
# ============================================================

plt.figure(figsize=(13, 7))

for operator in sorted(
    df["OPERATOR"].unique()
):

    subset = df[
        df["OPERATOR"] == operator
    ].sort_values("YEAR_START")

    plt.plot(
        subset["YEAR_START"],
        subset["FREIGHT_TOTAL"],
        marker="o",
        linewidth=2,
        label=operator
    )

plt.title(
    "AIS - Historical Freight Traffic Trends",
    fontsize=15,
    fontweight="bold"
)

plt.xlabel("Financial Year Starting")
plt.ylabel("Freight Traffic")

plt.legend()
plt.grid(alpha=0.3)

save_plot("ais_freight_trend_graph.png")

# ============================================================
# 33. SAVE AIS SUMMARY CSV
# ============================================================

summary = {
    "Algorithm": "AIS Optimized Random Forest",
    "AIS_Best_Validation_Error": best_fitness,
    "Training_Samples": len(X_train),
    "Testing_Samples": len(X_test),
    **ais_metrics
}

summary_df = pd.DataFrame([
    summary
])

summary_df.to_csv(
    ROOT / "ais_model_summary.csv",
    index=False
)

# ============================================================
# 34. VERIFY SAVED MODEL
# ============================================================

loaded_model = joblib.load(
    ROOT / "ais_model.pkl"
)

sample_input = X.tail(3)

sample_predictions = loaded_model.predict(
    sample_input
)

print("\nSaved PKL model verification:")
print(sample_predictions)

with h5py.File(
    ROOT / "ais_model.h5",
    "r"
) as h5:

    model_data = (
        h5["serialized_model"][:].tobytes()
    )

# Only load H5/pickle artifacts from trusted sources.

restored_model = pickle.loads(
    model_data
)

restored_predictions = restored_model.predict(
    sample_input
)

assert np.allclose(
    sample_predictions,
    restored_predictions
)

print("\nH5 verification successful!")

# ============================================================
# 35. DISPLAY ALL GENERATED FILES
# ============================================================

print("\n" + "=" * 70)
print("AIS PROJECT COMPLETED SUCCESSFULLY")
print("=" * 70)

print("\nOptimized hyperparameters:")
print(best_params)

print("\nAIS Test Metrics:")

for key, value in ais_metrics.items():
    print(f"{key}: {value:.4f}")

print("\nGenerated AIS files:")

files = sorted(
    ROOT.glob("ais_*")
)

for file in files:
    print("[SAVED]", file.name)

print("\nTotal AIS files:", len(files))

print("\nOutput folder:")
print(ROOT)

print("\nNOTE:")
print(
    "The dataset is small. AIS optimization "
    "and future forecasts are exploratory."
)
```
