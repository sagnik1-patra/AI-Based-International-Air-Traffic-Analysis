```python
import os
import warnings
from pathlib import Path

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import joblib

from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
from sklearn.impute import SimpleImputer

from sklearn.linear_model import LinearRegression, Ridge
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import (
    RandomForestRegressor,
    GradientBoostingRegressor
)
from sklearn.multioutput import MultiOutputRegressor

from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

warnings.filterwarnings("ignore")

# ============================================================
# STEP 1: PROJECT DIRECTORY
# ============================================================

PROJECT_DIR = Path(
    r"C:\Users\sagni\Downloads"
    r"\AI-Based International Air Traffic Analysis"
)

DATASET_PATH = PROJECT_DIR / "IntlTrafficData.csv"

PROJECT_DIR.mkdir(parents=True, exist_ok=True)

sns.set_theme(style="whitegrid")

plt.rcParams["figure.figsize"] = (12, 6)
plt.rcParams["font.size"] = 11

print("=" * 65)
print("AI-BASED INTERNATIONAL AIR TRAFFIC ANALYSIS")
print("=" * 65)

# ============================================================
# STEP 2: HELPER FUNCTION TO SAVE AND SHOW GRAPHS
# ============================================================

def save_show(filename):

    path = PROJECT_DIR / filename

    plt.tight_layout()

    plt.savefig(
        path,
        dpi=300,
        bbox_inches="tight"
    )

    print(f"\nGraph saved: {path}")

    plt.show()
    plt.close()

# ============================================================
# STEP 3: LOAD DATASET
# ============================================================

if not DATASET_PATH.exists():
    raise FileNotFoundError(
        f"Dataset not found: {DATASET_PATH}"
    )

df = pd.read_csv(DATASET_PATH)

print("\nDataset loaded successfully!")
print("Shape:", df.shape)

print("\nFirst five rows:")
print(df.head())

# ============================================================
# STEP 4: CLEAN DATASET
# ============================================================

df.columns = (
    df.columns
    .str.strip()
    .str.upper()
    .str.replace(r"\s+", "_", regex=True)
)

required_columns = [
    "YEAR",
    "OPERATOR",
    "PAX_TOTAL",
    "FREIGHT_TOTAL"
]

missing = [
    col for col in required_columns
    if col not in df.columns
]

if missing:
    raise ValueError(
        f"Required columns missing: {missing}"
    )

df["YEAR"] = df["YEAR"].astype(str).str.strip()

df["OPERATOR"] = (
    df["OPERATOR"]
    .astype(str)
    .str.strip()
    .str.upper()
)

# Convert financial year into starting year
# Example: 2010-11 -> 2010

df["YEAR_START"] = pd.to_numeric(
    df["YEAR"].str.extract(r"(\d{4})")[0],
    errors="coerce"
)

numeric_columns = [
    "PAX_TO",
    "PAX_FROM",
    "PAX_TOTAL",
    "FREIGHT_TO",
    "FREIGHT_FROM",
    "FREIGHT_TOTAL"
]

for col in numeric_columns:

    if col in df.columns:

        df[col] = pd.to_numeric(
            df[col].astype(str).str.replace(
                ",", "", regex=False
            ),
            errors="coerce"
        )

df = df.dropna(
    subset=[
        "YEAR_START",
        "OPERATOR",
        "PAX_TOTAL",
        "FREIGHT_TOTAL"
    ]
).copy()

df["YEAR_START"] = df["YEAR_START"].astype(int)

df = df.sort_values(
    ["YEAR_START", "OPERATOR"]
).reset_index(drop=True)

print("\nCleaned dataset:")
print(df.head())

print("\nMissing values:")
print(df.isnull().sum())

# ============================================================
# STEP 5: CORRELATION HEATMAP
# ============================================================

print("\nGenerating correlation heatmap...")

heatmap_columns = [
    col for col in numeric_columns
    if col in df.columns
]

heatmap_data = df[
    heatmap_columns + ["YEAR_START"]
].select_dtypes(include=[np.number])

correlation_matrix = heatmap_data.corr()

plt.figure(figsize=(12, 9))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    linewidths=0.5,
    square=True
)

plt.title(
    "International Air Traffic Correlation Heatmap",
    fontsize=15,
    fontweight="bold"
)

save_show("heatmap.png")

# ============================================================
# STEP 6: SELECT FEATURES AND TARGETS
# ============================================================

# Use year and operator to predict traffic.
# Do not use PAX_TO / PAX_FROM to predict
# PAX_TOTAL because that would cause leakage.

feature_columns = [
    "YEAR_START",
    "OPERATOR"
]

target_columns = [
    "PAX_TOTAL",
    "FREIGHT_TOTAL"
]

X = df[feature_columns].copy()
y = df[target_columns].copy()

# ============================================================
# STEP 7: TIME-AWARE TRAIN/TEST SPLIT
# ============================================================

# Important: All records from a given year
# must remain in the same split.

years = sorted(df["YEAR_START"].unique())

if len(years) < 3:
    raise ValueError(
        "At least 3 different years are required."
    )

split_index = max(
    2,
    int(len(years) * 0.70)
)

split_index = min(
    split_index,
    len(years) - 1
)

train_years = years[:split_index]
test_years = years[split_index:]

train_mask = df["YEAR_START"].isin(train_years)
test_mask = df["YEAR_START"].isin(test_years)

X_train = X.loc[train_mask].copy()
X_test = X.loc[test_mask].copy()

y_train = y.loc[train_mask].copy()
y_test = y.loc[test_mask].copy()

print("\nTraining years:", train_years)
print("Testing years:", test_years)

print("Training samples:", len(X_train))
print("Testing samples:", len(X_test))

# ============================================================
# STEP 8: PREPROCESSOR
# ============================================================

def create_preprocessor():

    numeric_pipeline = Pipeline([
        (
            "imputer",
            SimpleImputer(strategy="median")
        )
    ])

    categorical_pipeline = Pipeline([
        (
            "imputer",
            SimpleImputer(strategy="most_frequent")
        ),
        (
            "encoder",
            OneHotEncoder(handle_unknown="ignore")
        )
    ])

    return ColumnTransformer([
        (
            "numeric",
            numeric_pipeline,
            ["YEAR_START"]
        ),
        (
            "categorical",
            categorical_pipeline,
            ["OPERATOR"]
        )
    ])

# ============================================================
# STEP 9: DEFINE MACHINE LEARNING MODELS
# ============================================================

models = {
    "Linear Regression": LinearRegression(),

    "Ridge Regression": Ridge(alpha=1.0),

    "Decision Tree": DecisionTreeRegressor(
        max_depth=3,
        random_state=42
    ),

    "Random Forest": RandomForestRegressor(
        n_estimators=100,
        max_depth=4,
        random_state=42
    ),

    "Gradient Boosting": MultiOutputRegressor(
        GradientBoostingRegressor(
            n_estimators=100,
            learning_rate=0.05,
            max_depth=2,
            random_state=42
        )
    )
}

# ============================================================
# STEP 10: TRAIN ALL MODELS
# ============================================================

results = []
trained_models = {}

for name, estimator in models.items():

    print("\nTraining:", name)

    pipeline = Pipeline([
        (
            "preprocessor",
            create_preprocessor()
        ),
        (
            "model",
            estimator
        )
    ])

    pipeline.fit(X_train, y_train)

    y_pred = pipeline.predict(X_test)

    y_pred = np.maximum(y_pred, 0)

    # Passenger metrics

    passenger_mae = mean_absolute_error(
        y_test["PAX_TOTAL"],
        y_pred[:, 0]
    )

    passenger_rmse = np.sqrt(
        mean_squared_error(
            y_test["PAX_TOTAL"],
            y_pred[:, 0]
        )
    )

    passenger_r2 = r2_score(
        y_test["PAX_TOTAL"],
        y_pred[:, 0]
    )

    # Freight metrics

    freight_mae = mean_absolute_error(
        y_test["FREIGHT_TOTAL"],
        y_pred[:, 1]
    )

    freight_rmse = np.sqrt(
        mean_squared_error(
            y_test["FREIGHT_TOTAL"],
            y_pred[:, 1]
        )
    )

    freight_r2 = r2_score(
        y_test["FREIGHT_TOTAL"],
        y_pred[:, 1]
    )

    # Normalize errors to compare two
    # targets with different units.

    passenger_scale = max(
        abs(y_train["PAX_TOTAL"].mean()),
        1.0
    )

    freight_scale = max(
        abs(y_train["FREIGHT_TOTAL"].mean()),
        1.0
    )

    normalized_error = (
        passenger_mae / passenger_scale
        + freight_mae / freight_scale
    ) / 2

    results.append({
        "Model": name,
        "Passenger_MAE": passenger_mae,
        "Passenger_RMSE": passenger_rmse,
        "Passenger_R2": passenger_r2,
        "Freight_MAE": freight_mae,
        "Freight_RMSE": freight_rmse,
        "Freight_R2": freight_r2,
        "Normalized_Error": normalized_error
    })

    trained_models[name] = pipeline

    print(
        "Passenger R2:",
        round(passenger_r2, 4)
    )

    print(
        "Freight R2:",
        round(freight_r2, 4)
    )

# ============================================================
# STEP 11: MODEL COMPARISON CSV
# ============================================================

comparison_df = pd.DataFrame(results)

comparison_df = comparison_df.sort_values(
    "Normalized_Error"
).reset_index(drop=True)

comparison_df.to_csv(
    PROJECT_DIR / "model_comparison.csv",
    index=False
)

print("\nModel comparison:")
print(comparison_df.to_string(index=False))

# ============================================================
# STEP 12: ACCURACY GRAPH (R2 SCORE)
# ============================================================

# R2 is a regression metric.
# It may be negative and is NOT a percentage.

print("\nGenerating accuracy graph...")

plot_df = comparison_df.set_index("Model")[
    ["Passenger_R2", "Freight_R2"]
]

ax = plot_df.plot(
    kind="bar",
    figsize=(13, 7),
    width=0.75
)

plt.title(
    "Model Accuracy Comparison (R2 Score)",
    fontsize=15,
    fontweight="bold"
)

plt.xlabel("Machine Learning Models")
plt.ylabel("R2 Score")

plt.xticks(rotation=30, ha="right")

plt.axhline(
    y=0,
    color="black",
    linewidth=1
)

plt.legend([
    "Passenger R2",
    "Freight R2"
])

plt.grid(axis="y", alpha=0.3)

save_show("accuracy_graph.png")

# ============================================================
# STEP 13: MODEL COMPARISON GRAPH
# ============================================================

print("\nGenerating comparison graph...")

plt.figure(figsize=(13, 7))

sns.barplot(
    data=comparison_df,
    x="Model",
    y="Normalized_Error",
    color="steelblue"
)

plt.title(
    "Machine Learning Model Comparison",
    fontsize=15,
    fontweight="bold"
)

plt.xlabel("Machine Learning Models")

plt.ylabel(
    "Normalized Mean Absolute Error (Lower is Better)"
)

plt.xticks(rotation=30, ha="right")

plt.grid(axis="y", alpha=0.3)

save_show("comparison_graph.png")

# ============================================================
# STEP 14: SELECT BEST MODEL
# ============================================================

best_model_name = comparison_df.iloc[0]["Model"]

best_model = trained_models[best_model_name]

print("\nBest model:", best_model_name)

# ============================================================
# STEP 15: GENERATE RESULT CSV
# ============================================================

print("\nGenerating result CSV...")

test_predictions = best_model.predict(X_test)

test_predictions = np.maximum(
    test_predictions, 0
)

result_df = df.loc[
    test_mask,
    ["YEAR", "YEAR_START", "OPERATOR"]
].copy()

result_df["Actual_Passengers"] = (
    y_test["PAX_TOTAL"].values
)

result_df["Predicted_Passengers"] = (
    test_predictions[:, 0]
)

result_df["Actual_Freight"] = (
    y_test["FREIGHT_TOTAL"].values
)

result_df["Predicted_Freight"] = (
    test_predictions[:, 1]
)

result_df["Passenger_Error"] = (
    result_df["Actual_Passengers"]
    - result_df["Predicted_Passengers"]
)

result_df["Freight_Error"] = (
    result_df["Actual_Freight"]
    - result_df["Predicted_Freight"]
)

result_df["Passenger_Absolute_Error"] = (
    result_df["Passenger_Error"].abs()
)

result_df["Freight_Absolute_Error"] = (
    result_df["Freight_Error"].abs()
)

result_df = result_df.sort_values(
    ["YEAR_START", "OPERATOR"]
).reset_index(drop=True)

result_df.to_csv(
    PROJECT_DIR / "result.csv",
    index=False
)

print("\nResult CSV:")
print(result_df.to_string(index=False))

# ============================================================
# STEP 16: RESULT GRAPH
# ============================================================

print("\nGenerating result graph...")

# Use passenger results to visualize
# actual versus predicted values.

plt.figure(figsize=(13, 7))

x_positions = np.arange(len(result_df))

plt.plot(
    x_positions,
    result_df["Actual_Passengers"],
    marker="o",
    linewidth=2,
    label="Actual Passengers"
)

plt.plot(
    x_positions,
    result_df["Predicted_Passengers"],
    marker="s",
    linestyle="--",
    linewidth=2,
    label="Predicted Passengers"
)

labels = (
    result_df["YEAR"].astype(str)
    + " - "
    + result_df["OPERATOR"].astype(str)
)

plt.xticks(
    x_positions,
    labels,
    rotation=45,
    ha="right"
)

plt.title(
    "International Air Traffic - Actual vs Predicted",
    fontsize=15,
    fontweight="bold"
)

plt.xlabel("Financial Year and Operator")
plt.ylabel("Passenger Traffic")

plt.legend()
plt.grid(alpha=0.3)

save_show("result_graph.png")

# ============================================================
# STEP 17: REFIT MODEL ON COMPLETE DATASET
# ============================================================

# The model above was used for evaluation.
# For future predictions, refit on all historical data.

final_model = Pipeline([
    (
        "preprocessor",
        create_preprocessor()
    ),
    (
        "model",
        models[best_model_name]
    )
])

final_model.fit(X, y)

joblib.dump(
    final_model,
    PROJECT_DIR / "visualization_best_model.pkl"
)

# ============================================================
# STEP 18: GENERATE FUTURE PREDICTIONS
# ============================================================

print("\nGenerating future traffic predictions...")

last_year = int(df["YEAR_START"].max())

future_years = list(
    range(last_year + 1, last_year + 6)
)

# TOTAL is an aggregate category.
# Forecast Indian and Foreign separately
# when both are available, then sum them.

available_operators = set(
    df["OPERATOR"].unique()
)

if {"INDIAN", "FOREIGN"}.issubset(
    available_operators
):
    forecast_operators = [
        "INDIAN",
        "FOREIGN"
    ]

else:
    forecast_operators = sorted(
        op for op in available_operators
        if op != "TOTAL"
    )

    if not forecast_operators:
        forecast_operators = ["TOTAL"]

future_records = []

for year in future_years:

    for operator in forecast_operators:

        future_records.append({
            "YEAR_START": year,
            "OPERATOR": operator
        })

future_df = pd.DataFrame(future_records)

future_predictions = final_model.predict(
    future_df[feature_columns]
)

future_predictions = np.maximum(
    future_predictions, 0
)

future_df["Predicted_Passengers"] = (
    future_predictions[:, 0]
)

future_df["Predicted_Freight"] = (
    future_predictions[:, 1]
)

# Aggregate the individual operator forecasts.
# This avoids independently predicting TOTAL
# and then double-counting it.

if (
    "INDIAN" in forecast_operators
    and "FOREIGN" in forecast_operators
):

    total_forecasts = future_df.groupby(
        "YEAR_START",
        as_index=False
    )[
        [
            "Predicted_Passengers",
            "Predicted_Freight"
        ]
    ].sum()

    total_forecasts["OPERATOR"] = "TOTAL"

    future_df = pd.concat(
        [future_df, total_forecasts],
        ignore_index=True
    )

future_df["YEAR"] = future_df[
    "YEAR_START"
].apply(
    lambda year: (
        f"{year}-{str((year + 1) % 100).zfill(2)}"
    )
)

future_df = future_df[
    [
        "YEAR",
        "YEAR_START",
        "OPERATOR",
        "Predicted_Passengers",
        "Predicted_Freight"
    ]
]

future_df = future_df.sort_values(
    ["YEAR_START", "OPERATOR"]
).reset_index(drop=True)

future_df.to_csv(
    PROJECT_DIR / "prediction.csv",
    index=False
)

print("\nFuture prediction CSV:")
print(future_df.to_string(index=False))

# ============================================================
# STEP 19: PREDICTION GRAPH
# ============================================================

print("\nGenerating prediction graph...")

# Historical TOTAL traffic, if available.
# Otherwise aggregate the operator data.

historical_total = df[
    df["OPERATOR"] == "TOTAL"
].copy()

if historical_total.empty:

    historical_total = df.groupby(
        "YEAR_START",
        as_index=False
    )[
        ["PAX_TOTAL", "FREIGHT_TOTAL"]
    ].sum()

future_total = future_df[
    future_df["OPERATOR"] == "TOTAL"
].copy()

if future_total.empty:

    future_total = future_df.groupby(
        "YEAR_START",
        as_index=False
    )[
        [
            "Predicted_Passengers",
            "Predicted_Freight"
        ]
    ].sum()

plt.figure(figsize=(13, 7))

plt.plot(
    historical_total["YEAR_START"],
    historical_total["PAX_TOTAL"],
    marker="o",
    linewidth=2,
    label="Historical Passenger Traffic"
)

plt.plot(
    future_total["YEAR_START"],
    future_total["Predicted_Passengers"],
    marker="s",
    linestyle="--",
    linewidth=2,
    label="Forecast Passenger Traffic"
)

plt.axvline(
    x=last_year,
    color="gray",
    linestyle=":",
    label="Forecast Boundary"
)

plt.title(
    "International Air Passenger Traffic Forecast",
    fontsize=15,
    fontweight="bold"
)

plt.xlabel("Financial Year Starting")
plt.ylabel("Passenger Traffic")

plt.legend()
plt.grid(alpha=0.3)

save_show("prediction_graph.png")

# ============================================================
# STEP 20: ADDITIONAL FREIGHT PREDICTION GRAPH
# ============================================================

print("\nGenerating freight prediction graph...")

plt.figure(figsize=(13, 7))

plt.plot(
    historical_total["YEAR_START"],
    historical_total["FREIGHT_TOTAL"],
    marker="o",
    linewidth=2,
    label="Historical Freight Traffic"
)

plt.plot(
    future_total["YEAR_START"],
    future_total["Predicted_Freight"],
    marker="s",
    linestyle="--",
    linewidth=2,
    label="Forecast Freight Traffic"
)

plt.axvline(
    x=last_year,
    color="gray",
    linestyle=":",
    label="Forecast Boundary"
)

plt.title(
    "International Air Freight Traffic Forecast",
    fontsize=15,
    fontweight="bold"
)

plt.xlabel("Financial Year Starting")
plt.ylabel("Freight Traffic")

plt.legend()
plt.grid(alpha=0.3)

save_show("freight_prediction_graph.png")

# ============================================================
# STEP 21: ADDITIONAL PASSENGER TREND GRAPH
# ============================================================

print("\nGenerating passenger trend graph...")

plt.figure(figsize=(13, 7))

for operator in sorted(df["OPERATOR"].unique()):

    operator_df = df[
        df["OPERATOR"] == operator
    ].sort_values("YEAR_START")

    plt.plot(
        operator_df["YEAR_START"],
        operator_df["PAX_TOTAL"],
        marker="o",
        linewidth=2,
        label=operator
    )

plt.title(
    "International Passenger Traffic by Operator",
    fontsize=15,
    fontweight="bold"
)

plt.xlabel("Financial Year Starting")
plt.ylabel("Passenger Traffic")

plt.legend()
plt.grid(alpha=0.3)

save_show("passenger_trend_graph.png")

# ============================================================
# STEP 22: ADDITIONAL FREIGHT TREND GRAPH
# ============================================================

print("\nGenerating freight trend graph...")

plt.figure(figsize=(13, 7))

for operator in sorted(df["OPERATOR"].unique()):

    operator_df = df[
        df["OPERATOR"] == operator
    ].sort_values("YEAR_START")

    plt.plot(
        operator_df["YEAR_START"],
        operator_df["FREIGHT_TOTAL"],
        marker="o",
        linewidth=2,
        label=operator
    )

plt.title(
    "International Freight Traffic by Operator",
    fontsize=15,
    fontweight="bold"
)

plt.xlabel("Financial Year Starting")
plt.ylabel("Freight Traffic")

plt.legend()
plt.grid(alpha=0.3)

save_show("freight_trend_graph.png")

# ============================================================
# STEP 23: SAVE SUMMARY
# ============================================================

summary = pd.DataFrame({
    "Metric": [
        "Best Model",
        "Training Samples",
        "Testing Samples",
        "Passenger R2",
        "Freight R2",
        "Passenger MAE",
        "Freight MAE",
        "Normalized Error"
    ],
    "Value": [
        best_model_name,
        len(X_train),
        len(X_test),
        comparison_df.iloc[0]["Passenger_R2"],
        comparison_df.iloc[0]["Freight_R2"],
        comparison_df.iloc[0]["Passenger_MAE"],
        comparison_df.iloc[0]["Freight_MAE"],
        comparison_df.iloc[0]["Normalized_Error"]
    ]
})

summary.to_csv(
    PROJECT_DIR / "model_summary.csv",
    index=False
)

# ============================================================
# STEP 24: FINAL OUTPUT
# ============================================================

print("\n" + "=" * 65)
print("PROJECT COMPLETED SUCCESSFULLY")
print("=" * 65)

print("\nBest Machine Learning Model:")
print(best_model_name)

print("\nGenerated Files:")

output_files = [
    "accuracy_graph.png",
    "heatmap.png",
    "comparison_graph.png",
    "result.csv",
    "result_graph.png",
    "prediction.csv",
    "prediction_graph.png",
    "model_comparison.csv",
    "freight_prediction_graph.png",
    "passenger_trend_graph.png",
    "freight_trend_graph.png",
    "model_summary.csv",
    "visualization_best_model.pkl"
]

for filename in output_files:

    path = PROJECT_DIR / filename

    if path.exists():
        print("[SAVED]", filename)
    else:
        print("[MISSING]", filename)

print("\nAll outputs saved in:")
print(PROJECT_DIR)

print("\nIMPORTANT:")
print(
    "Forecasts are exploratory because "
    "the historical dataset is very small."
)
```
